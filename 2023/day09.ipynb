```python
import os
```

```python
file = 'data/day09.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
def extrap(nums):
    diff_lists = [nums]
    while True:
        diff = [int(nums[i+1]) - int(nums[i]) for i in range(len(nums) - 1)]
        diff_lists.append(diff)
        if all(x == 0 for x in diff_lists[-1]):
            break
        else:
            nums = diff_lists[-1]
    return sum(x[-1] for x in diff_lists)


def extrap_backwards(nums):
    diff_lists = [nums]
    while True:
        diff = [int(nums[i+1]) - int(nums[i]) for i in range(len(nums) - 1)]
        diff_lists.append(diff)
        if all(x == 0 for x in diff_lists[-1]):
            break
        else:
            nums = diff_lists[-1]
    ex_point = 0
    for x in diff_lists[::-1]:
        ex_point = x[0] - ex_point 
    return ex_point
```

```python
# part one
data = [[int(x) for x in line.split()] for line in lines]
print(sum(extrap(line) for line in data))
```

```python
# part two
print(sum(extrap_backwards(line) for line in data))
```
