```python
import os
```

```python
file = 'data/day11.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# expand universe
lines = [list(x) for x in zip(*lines)]
exp_col_list = [i for i, line in enumerate(lines) if set(line) == {'.'}]
lines = [list(x) for x in zip(*lines)]
exp_row_list = [i for i, line in enumerate(lines) if set(line) == {'.'}]
```

```python
# part one
galaxy_dict, count, exp_mult = {}, 0, 2 # register galaxies
for row, line in enumerate(lines): 
    for col, char in enumerate(line):
        if char == '#':
            row_offset = len([x for x in exp_row_list if x < row])
            row_offset = row_offset * exp_mult - row_offset
            col_offset = len([x for x in exp_col_list if x < col]) 
            col_offset = col_offset * exp_mult - col_offset
            galaxy_dict[count] = (row + row_offset, col + col_offset)
            count += 1

distance_dict, galaxy_qty = {}, len(galaxy_dict)
for i in range(galaxy_qty): # collate distances
    for j in range(i+1, galaxy_qty):
        g1x, g1y = galaxy_dict[i]
        g2x, g2y = galaxy_dict[j]
        distance_dict[(i, j)] = abs(g1x - g2x) + abs(g1y - g2y)
print(sum(distance_dict.values()))
```

```python
# part two
galaxy_dict, count, exp_mult = {}, 0, 1_000_000 # register galaxies
for row, line in enumerate(lines): 
    for col, char in enumerate(line):
        if char == '#':
            row_offset = len([x for x in exp_row_list if x < row])
            row_offset = row_offset * exp_mult - row_offset
            col_offset = len([x for x in exp_col_list if x < col]) 
            col_offset = col_offset * exp_mult - col_offset
            galaxy_dict[count] = (row + row_offset, col + col_offset)
            count += 1

distance_dict, galaxy_qty = {}, len(galaxy_dict)
for i in range(galaxy_qty): # collate distances
    for j in range(i+1, galaxy_qty):
        g1x, g1y = galaxy_dict[i]
        g2x, g2y = galaxy_dict[j]
        distance_dict[(i, j)] = abs(g1x - g2x) + abs(g1y - g2y)
print(sum(distance_dict.values()))
```
