```python
import os
```

```python
file = 'data/day04.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
card_dict = {}
for line in lines:
    card_number, remainder = line.split(': ')
    card_number = int(card_number.replace('Card', '').strip())
    my_nums, winning_nums = remainder.split(' | ')
    my_nums = my_nums.replace('  ', ' ')
    winning_nums = winning_nums.replace('  ', ' ')
    my_nums = [int(x) for x in my_nums.strip().split(' ')]
    winning_nums = [int(x) for x in winning_nums.strip().split(' ')]
    card_dict[card_number] = [my_nums, winning_nums]
```

```python
# part one
card_sum = 0
for card in card_dict:
    my_nums, winning_nums = card_dict[card]
    exp = len(set(my_nums).intersection(winning_nums))
    if exp:
        card_sum += 2 ** (len(set(my_nums).intersection(winning_nums)) - 1)
print(card_sum)
```

```python
# part two
score_dict = {}
for card in card_dict:
    my_nums, winning_nums = card_dict[card]
    score_dict[card] = len(set(my_nums).intersection(winning_nums))
new_scores = {num + 1: 1 for num in range(len(score_dict))}
for card in score_dict:
    for offset in range(score_dict[card]):
        new_scores[card + offset + 1] += new_scores[card]
print(sum(new_scores.values()))
```
