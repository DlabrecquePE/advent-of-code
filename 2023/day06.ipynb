```python
import os
import math
```

```python
file = 'data/day06.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
times = [int(x) for x in lines[0].strip('Time:').split()]
distances = [int(x) for x in lines[1].strip('Distance:').split()]

win_counts = []
for i, time in enumerate(times):
    win_count = 0
    for j in range(time):
        if j * (time - j) > distances[i]:
            win_count += 1
    win_counts.append(win_count)
print(math.prod(win_counts))
```

```python
time = int(lines[0].strip('Time:').replace(' ', ''))
distance = int(lines[1].strip('Distances:').replace(' ', ''))

high = (time + ((time ** 2) - 4 * distance) ** 0.5) // 2
low = (time - ((time ** 2) - 4 * distance) ** 0.5) // 2
print(int(high - low))  # lol close enough
```
