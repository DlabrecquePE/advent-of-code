```python
import os
```

```python
file = 'data/day01.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
total = 0
for line in lines:
    numerals = [char for char in line if char.isdigit()]
    total += int(numerals[0] + numerals[-1])
print(total)
```

```python
# part two
words = {'one' : '1', 'two': '2', 'three': '3', 'four': '4', 'five': '5', 
         'six': '6', 'seven': '7', 'eight': '8', 'nine': '9'}
total = 0
for line in lines:
    first_num = last_num = first = last = 0
    for i, char in enumerate(line):
        if char.isdigit():
            first_num_index, first_num = i, char
            break
    for i, char in enumerate(line[::-1]):
        if char.isdigit():
            last_num_index, last_num = i, char
            break
    first_prefix = line[:first_num_index] if first_num else line
    last_suffix = line[len(line) - last_num_index:] if last_num else line
    for i in range(len(first_prefix)):
        if first:
            break
        for word in words:
            if first_prefix[i:].startswith(word):
                first = words[word]
                break
    if not first:
        first = first_num
    while last_suffix:
        if last:
            break
        for word in words:
            if last_suffix.endswith(word):
                last = words[word]
                break
        last_suffix = last_suffix[:-1]
    if not last:
        last = last_num
    total += int(first + last)
print(total)
    
```
