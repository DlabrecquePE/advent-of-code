```python
import os
```

```python
file = 'data/day08.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
from math import gcd
def lcm(nums):
    lcm = 1
    for i in nums:
        lcm = lcm*i//gcd(lcm, i)
    return lcm
```

```python
instructions = lines.pop(0)
lines.pop(0)
map_dict = {}
for line in lines:
    key, parse = line.split(' = (')
    left, right = parse.rstrip(')').split(', ')
    map_dict[key] = (left, right)
```

```python
# part one
start_node = 'AAA'
current_node, step_count, end_node_found = start_node, 0, False
while True:
    if end_node_found:
        break
    for direction in instructions:
        step_count += 1
        left_node, right_node = map_dict[current_node]
        current_node = left_node if direction == 'L' else right_node
        if current_node == 'ZZZ':
            print(step_count)
            end_node_found = True
            break
```

```python
# part two
start_nodes = [node for node in map_dict.keys() if node.endswith('Z')]
end_nodes_found = False
step_counts = []
for node in start_nodes:
    step_count, end_node_found, current_node = 0, False, node
    while True:
        if end_node_found:
            break
        for direction in instructions:
            step_count += 1
            left_node, right_node = map_dict[current_node]
            current_node = left_node if direction == 'L' else right_node
            if current_node.endswith('Z'):
                step_counts.append(step_count)
                end_node_found = True
                break
print(lcm(step_counts)) # damn, this should not work.
```
