```python
import os
```

```python
file = 'data/day15.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = fp.readlines()
```

```python
def day15_hash(hash_code):
    current_value = 0
    for char in hash_code:
        ascii_value = ord(char)
        current_value += ascii_value
        current_value *= 17
        current_value %= 256
    return current_value
```

```python
# part one
lines = list(lines[0].split(','))
print(sum([day15_hash(x) for x in lines]))
```

```python
# part two
box_dict = {box: [] for box in range(256)}
for line in lines:
    label = line[:-1] if '-' in line else line[:-2]
    box = day15_hash(label)
    if line[-2] == '=':
        focal_length = int(line[-1])
        if label in [x[0] for x in box_dict[box]]:
            for i, pair in enumerate(box_dict[box]):
                if pair[0] == label:
                    box_dict[box][i] = (label, focal_length)
        else:
            box_dict[box].append((label, focal_length))
    elif [True for x in box_dict[box] if x[0] == label]:
        for x in box_dict[box].copy():
            if x[0] == label:
                box_dict[box].remove(x)
focusing_power = 0
for box in box_dict:
    for slot, lense in enumerate(box_dict[box]):
        focusing_power += (1+box) * (1+slot) * lense[1]
print(focusing_power)
```
