```python
import os
```

```python
file = 'data/test.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
dig_plan = {}
for step, line in enumerate(lines):
    direct, dist, color = line.split(' ')
    dig_plan[step] = [direct, int(dist), color[1:-1]]
```

```python
curr = (0, 0)
delta_map = {'U': (-1, 0), 'D': (1, 0), 'L': (0, -1), 'R': (0, 1)}
trench_map = set()
for step in dig_plan:
    row, col = curr
    direct, dist, color = dig_plan[step]
    d_row, d_col = delta_map[direct]
    trench_coords = {(row + d_row*i, col + d_col*i) for i in range(dist)}
                     
    trench_map = trench_map.union(trench_coords)
                
```
