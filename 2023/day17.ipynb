```python
import os
```

```python
file = 'data/test.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
def dijkstra(graph, source, target):
    ''' Following Pseudocode from 
    https://en.wikipedia.org/wiki/Dijkstra%27s_algorithm''' 
    
    dist, prev, Q, graph_edges = {}, {}, [], {}
    row_max, col_max = len(graph), len(graph[0])
    for row, line in enumerate(graph):
        for col, char in enumerate(line):
            v = (row, col)
            dist[v] = float('inf')
            prev[v] = None
            graph_edges[v] = int(char)
            Q.append(v)
    dist[source] = 0
    
    while Q:
        u = [y for y in Q if dist[y] == min([dist[x] for x in Q])][0]
        if u == target:
            break
        Q.remove(u)
        row, col, neighbors = *u, []
        n, s, e, w = (row-1,col), (row+1,col), (row,col+1), (row,col-1)
        if 0 < row and n in Q:
            neighbors.append(n)
        if 0 < col and w in Q:
            neighbors.append(w)
        if row < row_max - 2 and s in Q:
            neighbors.append(s)
        if col < col_max - 2 and e in Q:
            neighbors.append(e)
                
        for v in neighbors:
            alt = dist[u] + graph_edges[v]
            if alt < dist[v]:
                dist[v] = alt
                prev[v] = u
        
    return dist, prev
```

```python
source, u, S = (0, 0), (len(lines) - 1, len(lines[0]) - 1), []
dist, prev = dijkstra(lines, source, u)
if prev[u] or u == source:
    while u:
        S = [u] + S
        u = prev[u]
print(S)
print(dist)
print(prev)
```
