```python
import os
```

```python
file = 'data/day16.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [list(x.rstrip()) for x in fp.readlines()]
```

```python
# part one
start_pos, start_vel = (0,0), (0,1)
row_max, col_max = len(lines), len(lines[0])
beam_stack, paths = [(start_pos, start_vel)], []
while beam_stack:
    pos, vel = beam_stack.pop()
    row, col = pos
    u_d, l_r = vel
    if not (0 <= row < row_max and 0 <= col < col_max) or (pos, vel) in paths:
        continue
    paths.append((pos, vel))
    match lines[row][col]:
        case '.':
            pos = (row + u_d, col) if u_d else (row, col + l_r)
        case '|':
            if l_r:
                beam_stack.append(((row - 1, col), (-1, 0)))
                pos = (row + 1, col)
                vel = (1, 0)
            else:
                pos = (row + u_d, col)
        case '-':
            if u_d:
                beam_stack.append(((row, col -1), (0, -1)))
                pos = (row, col + 1)
                vel = (0, 1)
            else:
                pos = (row, col + l_r)
        case '/':
            if l_r:
                pos = (row - l_r, col)
                vel = (-l_r, 0)
            else:
                pos = (row, col - u_d)
                vel = (0, -u_d)
        case '\\':
            if l_r:
                pos = (row + l_r, col)
                vel = (l_r, 0)
            else:
                pos = (row, col + u_d)
                vel = (0, u_d)
    beam_stack.append((pos, vel))
print(len(set(x[0] for x in paths)))
```
