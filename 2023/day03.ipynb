```python
import os
```

```python
file = 'data/day03.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
max_rows, max_cols = len(lines), len(lines[0])
```

```python
# part one
symbol_coords = []
for row, line in enumerate(lines):
    for col, char in enumerate(line):
        if not char.isnumeric() and char != '.':
            symbol_coords.append((row, col))
valid_coords = []
for coord in symbol_coords:
        row, col = coord
        if row + 1 <= max_rows and col + 1 <= max_cols and lines[row + 1][col + 1].isnumeric():
            valid_coords.append((row + 1, col + 1))
        if row - 1 >= 0 and col - 1 >= 0 and lines[row - 1][col - 1].isnumeric():
            valid_coords.append((row - 1, col - 1))
        if row + 1 <= max_rows and col - 1 >= 0 and lines[row + 1][col - 1].isnumeric():
            valid_coords.append((row + 1, col - 1))
        if row - 1 >= 0 and col + 1 <= max_cols and lines[row - 1][col + 1].isnumeric():
            valid_coords.append((row - 1, col + 1))
        if col + 1 <= max_cols and lines[row][col + 1].isnumeric():
            valid_coords.append((row, col + 1))
        if col - 1 >= 0 and lines[row][col - 1].isnumeric():
            valid_coords.append((row, col - 1))
        if row + 1 <= max_rows and lines[row + 1][col].isnumeric():
            valid_coords.append((row + 1, col))
        if row - 1 >= 0 and lines[row - 1][col].isnumeric():
            valid_coords.append((row - 1, col)) 
part_coords = []
for coord in valid_coords:
    row, col = coord
    while True:
        if col > 0 and lines[row][col - 1].isnumeric():
            col -= 1
        else:
            part_coords.append((row, col))
            break
number_list = []
for coord in set(part_coords):
    row, col = coord
    number = ''
    while True:
        if col < max_cols and lines[row][col].isnumeric():
            number += lines[row][col]
            col += 1
        else:
            break
    number_list.append(int(number))
print(sum(number_list))
            
```

```python
# part two
symbol_coords = []
for row, line in enumerate(lines):
    for col, char in enumerate(line):
        if char == '*':
            symbol_coords.append((row, col))
max_rows, max_cols = len(lines), len(lines[0])
gear_dict = {}
for coord in symbol_coords:
        row, col = coord
        valid_coords = []
        if row + 1 <= max_rows and col + 1 <= max_cols and lines[row + 1][col + 1].isnumeric():
            valid_coords.append((row + 1, col + 1))
        if row - 1 >= 0 and col - 1 >= 0 and lines[row - 1][col - 1].isnumeric():
            valid_coords.append((row - 1, col - 1))
        if row + 1 <= max_rows and col - 1 >= 0 and lines[row + 1][col - 1].isnumeric():
            valid_coords.append((row + 1, col - 1))
        if row - 1 >= 0 and col + 1 <= max_cols and lines[row - 1][col + 1].isnumeric():
            valid_coords.append((row - 1, col + 1))
        if col + 1 <= max_cols and lines[row][col + 1].isnumeric():
            valid_coords.append((row, col + 1))
        if col - 1 >= 0 and lines[row][col - 1].isnumeric():
            valid_coords.append((row, col - 1))
        if row + 1 <= max_rows and lines[row + 1][col].isnumeric():
            valid_coords.append((row + 1, col))
        if row - 1 >= 0 and lines[row - 1][col].isnumeric():
            valid_coords.append((row - 1, col))
        gear_dict[coord] = valid_coords
valid_parts, number_list = [], []
for gear_coord in gear_dict:
    part_coords = []
    for coord in gear_dict[gear_coord]:
        row, col = coord
        while True:
            if col > 0 and lines[row][col - 1].isnumeric():
                col -= 1
            else:
                part_coords.append((row, col))
                break
    part_coords = set(part_coords)
    if len(part_coords) == 2:
        pair = []
        for coord in part_coords:
            row, col = coord
            number = ''
            while True:
                if col < max_cols and lines[row][col].isnumeric():
                    number += lines[row][col]
                    col += 1
                else:
                    break
            pair.append(int(number))
        number_list.append(pair[0] * pair[1])
print(sum(number_list))
```
