```python
import os
```

```python
file = 'data/day12.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
def arrangements(springs, groups):
    while springs and springs[0] == '.':
        springs.pop(0)
    if not springs:
        return 1 if not groups else 0
    if not groups:
        return 1 if "#" not in springs else 0
    if len(springs) == 1:
        return 1 if groups == [1] and springs != ['.'] else 0
    # match springs[0]:
    #     case '.':
    #         return arrangements(springs[1:], groups)
    if springs[0] == '#':
        # case '#':
        g_head, look_ahead = groups[0], springs[1]
        if g_head == 1 and look_ahead != '#':
            return arrangements(springs[2:], groups[1:])
        elif g_head != 1 and look_ahead != '.':
            return arrangements(['#'] + springs[2:], [g_head - 1] + groups[1:])
        else:
            return 0
    else:
        # case '?':
        g_head, look_ahead = groups[0], springs[1]
        if g_head == 1 and look_ahead != '#':
            branch1 = arrangements(springs[2:], groups[1:])
        elif g_head != 1 and look_ahead != '.':
            branch1 = arrangements(['#'] + springs[2:], [g_head - 1] + groups[1:])
        else:
            branch1 = 0
        branch2 = arrangements(springs[1:], groups)
        return branch1 + branch2
```

```python
# part one
count = 0
for line in lines:
    spring_code, groups = line.split(' ')
    spring_code = list(spring_code)
    groups = [int(x) for x in groups.split(',')]
    count += arrangements(spring_code, groups)
print(count)
```

```python
# part two
count = 0
for line in lines:
    spring_code, groups = line.split(' ')
    spring_code = list('?'.join(spring_code for _ in range(5)))
    groups = [int(x) for x in groups.split(',')] * 5
    count += arrangements(spring_code, groups)
print(count)
```
