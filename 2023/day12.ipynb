```python
file = 'data/test.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
def arments(data1, data2):
    springs, groups = data1, data2
    if not springs:
        return 1 if not groups else 0
    if not groups:
        return 1 if "#" not in springs else 0
    s_head, g_head = springs[0], groups[0]
    if len(springs) == 1:
        return 1 if groups == [1] and springs != ['.'] else 0
    match springs[0] + springs[1]:
        case '..':
            return arments(springs[2:], groups)
        case '.#':
            return arments(springs[1:], groups)
        case '##':
            if groups[0] != 1:
                groups[0] -= 1
                return arments(springs[1:], groups)
            else:
                return 0 
        case '#.':
            if groups[0] == 1:
                return arments(springs[2:], groups[1:])
            else:
                return 0
        case '??':
            branch1 = arments(['#'] + springs[1:], groups)
            branch2 = arments(['.'] + springs[1:], groups)
            return branch1 + branch2
        case '.?':
            return arments(springs[1:], groups)
        case '?.':
            branch1 = arments(['#'] + springs[1:], groups)
            branch2 = arments(['.'] + springs[1:], groups)
            return branch1 + branch2
        case '?#':
            branch1 = arments(['#'] + springs[1:], groups)
            branch2 = arments(['.'] + springs[1:], groups)
            return branch1 + branch2
        case '#?':
            branch1 = arments(['#'] + ['#'] + springs[2:], groups)
            branch2 = arments(['#'] + ['.'] + springs[2:], groups)
```

```python
spring_code = list('.??..??...?##.')
groups = [1,1,3]
print(arments(spring_code, groups))
```
