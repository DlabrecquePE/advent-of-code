```python
import os
```

```python
file = 'data/day13.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
pattern, pattern_dict, count = [], {}, 0
for line in lines:
    if not line or line == '':
        pattern_dict[count] = pattern
        pattern = []
        count += 1
    else:
        pattern.append(line)
pattern_dict[count] = pattern
```

```python
# part one
total, reflection_dict = 0, {}
for data in pattern_dict:
    pattern1 = list(pattern_dict[data])
    pattern2 = [''.join(x) for x in list(zip(*pattern1))]
    for vertical, pattern in enumerate([pattern1, pattern2]):
        for i in range(len(pattern[0]) - 1):
            a = [line[0:i+1] for line in pattern]
            b = [line[2*i+1:i:-1] for line in pattern]
            comp_range = min(len(a[0]), len(b[0]))
            a = [x[-comp_range:] for x in a]
            b = [x[:comp_range] for x in b]
            if a == b:
                reflection_dict[data] = (i, vertical)
                total += i + 1 if not vertical else (i+1) * 100
print(total)
```

```python
# part two
total = 0
for original in pattern_dict:
    pattern_list, reflection_found = [], False
    op = pattern_dict[original]
    for row, line in enumerate(op):
        for col, char in enumerate(line):
            new_char = '#' if char == '.' else '.'
            new_line = line[:col] + new_char + line[col + 1:]
            pattern_copy = op[:row] + [new_line] + op[row + 1:]
            pattern_list.append(pattern_copy)
    for pattern in pattern_list:
        pattern1 = list(pattern)
        pattern2 = [''.join(x) for x in list(zip(*pattern1))]
        for vertical, pattern in enumerate([pattern1, pattern2]):
            if reflection_found == True:
                break
            for i in range(len(pattern[0]) - 1):
                a = [line[0:i+1] for line in pattern]
                b = [line[2*i+1:i:-1] for line in pattern]
                comp_range = min(len(a[0]), len(b[0]))
                a = [x[-comp_range:] for x in a]
                b = [x[:comp_range] for x in b]
                if a == b and reflection_dict[original] != (i, vertical):
                    total += i + 1 if not vertical else (i+1) * 100
                    reflection_found = True
                    break
print(total)                
```
