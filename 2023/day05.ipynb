```python
import os
```

```python
file = 'data/day05.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
initial_seed_list = [int(x) for x in lines.pop(0).lstrip('seeds: ').split(' ')]
map_dict = {}
count = 0
lines.pop(0)
lines.pop(0)
while lines:
    current_line = lines.pop(0)
    map_data = []
    while current_line:
        map_data.append(tuple(int(x) for x in current_line.split(' ')))
        if lines:
            current_line = lines.pop(0)
        else:
            break
    map_dict[count] = map_data
    count += 1
    if lines:
        current_line = lines.pop(0)
    else:
        break
```

```python
# part one
seed_list = initial_seed_list.copy()

for seed_map in map_dict:
    seeds = seed_list.copy()
    for elem in map_dict[seed_map]:
        dest, source, length = elem
        for i, seed in enumerate(seeds):
            if seed >= source and seed < source + length:
                seed_list[i] = dest + seed - source
print(min(seed_list))
```

```python
# part two
old_seed_list = initial_seed_list.copy()
seed_list = []
while old_seed_list:
    seed_list.append((old_seed_list.pop(0), old_seed_list.pop(0)))

for seed_map in map_dict:    
    new_seed_list = []
    while seed_list:
        seed = seed_list.pop()
        start, span = seed
        mappings_checked, total_mappings = 0, len(map_dict[seed_map])
        for elem in map_dict[seed_map]:
            dest, source, length = elem
            diff = abs(start - source)
                    
            if start >= source and start + span <= source + length:
                # mapping entirely encompasses seed range
                # seed   ----
                # map  --------  
                new_seed = (dest + diff, span)
                if new_seed not in new_seed_list:
                    new_seed_list.append(new_seed)
                
            elif start < source and source < start + span <= source + length:
                # mapping does not cover left of seed range
                # seed ------
                # map    ------
                new_seed = (dest, span - diff)
                prefix = (start, diff)
                if new_seed not in new_seed_list:
                    new_seed_list.append(new_seed)
                if prefix not in seed_list:
                    seed_list.append(prefix)
                
            elif start >= source and start + span >= source + length > start:
                # mapping does not cover right of seed range
                # seed   ------
                # map  ------
                new_seed = (dest + diff, length - diff)
                suffix = (source + length, span + diff - length)
                if new_seed not in new_seed_list:
                    new_seed_list.append(new_seed)
                if suffix not in seed_list:
                    seed_list.append(suffix)
                
            elif start < source and start + span > source + length:
                # mapping encompassed by seed range
                # seed --------
                # map    ----
                new_seed = (dest, length)
                prefix = (start, diff)
                suffix = (source + length, span + diff - length)
                if new_seed in new_seed_list:
                    new_seed_list.append(new_seed)
                if prefix in seed_list:
                    seed_list.append(prefix)
                if suffix in seed_list:
                    seed_list.append(suffix)
            else:
                mappings_checked += 1
        if mappings_checked == total_mappings and seed not in new_seed_list:
            new_seed_list.append(seed)
    seed_list = new_seed_list
print(min(seed[0] for seed in seed_list))
```
