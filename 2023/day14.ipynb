```python
import os
```

```python
file = 'data/day14.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
grid = {}
for row, line in enumerate(lines):
    for col, char in enumerate(line):
        grid[(row, col)] = char
```

```python
old_grid = grid.copy()
while True: # moves all round boulders north
    new_grid = old_grid.copy()
    for coord in old_grid:
        row, col = coord
        char = old_grid[coord]
        north = (row - 1, col)
        if row > 0 and char == 'O' and old_grid[north] == '.':
            new_grid[coord] = '.'
            new_grid[north] = 'O'
    if new_grid == old_grid:
        break
    old_grid = new_grid
```

```python
# part one
total_rows, total_load = len(lines), 0
for coord in new_grid:
    if new_grid[coord] == 'O':
        total_load += total_rows - coord[0]
print(total_load)
```
