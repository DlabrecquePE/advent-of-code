```python
import os
import re
```

```python
file = 'data/day02.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
cube_max = {'red': 12, 'green': 13, 'blue': 14}
game_dict = {}
for line in lines:
    game_id, tail = line.split(': ')
    game_id = int(re.sub("[^0-9^.]", "", game_id))
    game_sets = tail.split('; ')
    games = []
    for game_set in game_sets:
        game = game_set.split(', ')
        game_round = {}
        for g in game:
            num, color = g.split(' ')
            game_round[color] = int(num)
        games.append(game_round)
    game_dict[game_id] = games
```

```python
# part one
valid_games = power_sum = 0
for game_id in game_dict:
    invalid = False
    for turn in game_dict[game_id]:
        if any(turn[color] > cube_max[color] for color in turn):
            invalid = True
    if not invalid:
        valid_games += game_id
print(valid_games)
```

```python
# part two
for game_id in game_dict:
    reds, greens, blues = ([] for _ in range(3))
    for turn in game_dict[game_id]:
        for color in turn:
            match color:
                case 'red':
                    reds.append(turn[color])
                case 'green':
                    greens.append(turn[color])
                case 'blue':
                    blues.append(turn[color])
    red = max(reds) if reds else 0
    green = max(greens) if greens else 0
    blue = max(blues) if blues else 0
    power_sum += red * green * blue
print(power_sum)
```
