```python
import os
```

```python
file = 'data/day07.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
from collections import Counter
def part_one_ranker(hand):
    card_values = {'A': 14, 'K': 13, 'Q': 12, 'J': 11, 'T': 10, '9': 9, '8': 8,
                   '7': 7, '6': 6, '5': 5, '4': 4, '3': 3, '2': 2}
    unique_cards = len(set(hand))
    if unique_cards in [5, 4, 1]:
        score = (14 - 2*unique_cards) * 10**12
    elif unique_cards == 3:
        if max(Counter(hand).values()) == 2:
            score = 7 * 10**12
        else:
            score = 8 * 10**12
    else:
        if max(Counter(hand).values()) == 3:
            score = 9 * 10**12
        else:
            score = 10 * 10**12
    for i, card in enumerate(hand):
        score += 10**(10 - 2*i) * card_values[card]
    return score


def part_two_ranker(hand):
    card_values = {'A': 14, 'K': 13, 'Q': 12, 'J': 1, 'T': 10, '9': 9, '8': 8,
                   '7': 7, '6': 6, '5': 5, '4': 4, '3': 3, '2': 2}
    wildcards = 'AKQT98765432'
    if 'J' in hand:
        score = max(part_two_ranker(hand.replace('J', x)) for x in wildcards)
        score = (score // 10**12) * 10**12
    else:
        unique_cards = len(set(hand))
        if unique_cards in [5, 4, 1]:
            score = (14 - 2*unique_cards) * 10**12
        elif unique_cards == 3:
            if max(Counter(hand).values()) == 2:
                score = 7 * 10**12
            else:
                score = 8 * 10**12
        else:
            if max(Counter(hand).values()) == 3:
                score = 9 * 10**12
            else:
                score = 10 * 10**12
    for i, card in enumerate(hand):
        score += 10**(10 - 2*i) * card_values[card]
    return score
    
```

```python
# part one
hands = [tuple(line.split()) for line in lines]
hand_rankings = {part_one_ranker(hand[0]): hand[1] for hand in hands}
winnings = []
for i, hand in enumerate(sorted(hand_rankings.keys())):
    winnings.append((i + 1) * int(hand_rankings[hand]))
print(sum(winnings))
```

```python
hand_rankings = {part_two_ranker(hand[0]): hand[1] for hand in hands}
winnings = []
for i, hand in enumerate(sorted(hand_rankings.keys())):
    winnings.append((i + 1) * int(hand_rankings[hand]))
print(sum(winnings))
```
