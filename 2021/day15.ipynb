```python
import os
```

```python
def dijkstra_grid(data):
    """ Dijkstra's algorithm for rectilinear grids of integers.
        
    This function takes a list of lists of integers as the grid, 
    and returns the smallest sum of those integers from the upper left
    grid corner to the lower right."""
    max_rows, max_cols = len(data), len(data[0])
    grid, distances, unvisited_set = {}, {}, set()
    for i in range(max_rows):
        for j in range(max_cols):
            grid[(i, j)] = data[i][j]
            distances[(i, j)] = float('inf')
            unvisited_set.add((i, j))
    start_node, final_node = (0, 0), (max_rows - 1, max_cols - 1)
    touched_set, current_node = {start_node}, start_node
    distances[start_node] = 0
    while True:
        neighbors, i, j = [], *current_node
        if i > 0:
            neighbors.append((i - 1, j))
        if j > 0:
            neighbors.append((i, j - 1))
        if i < max_rows - 1:
            neighbors.append((i + 1, j))
        if j < max_cols - 1:
            neighbors.append((i, j + 1))
        for neighbor in neighbors:
            if neighbor in unvisited_set:
                current_distance = distances[current_node]
                neighbor_distance = grid[neighbor]
                tentative_distance = current_distance + neighbor_distance
                assigned_distance = distances[neighbor]
                if tentative_distance < assigned_distance:
                    distances[neighbor] = tentative_distance
                    touched_set.add(neighbor)
        unvisited_set.remove(current_node)
        touched_set.remove(current_node)
        if final_node not in unvisited_set:
            break
        shortest_paths = {node: distances[node] for node in touched_set}
        current_node = min(shortest_paths, key=shortest_paths.get)
    return distances[final_node]
```

```python
file = 'data/day15.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
cave = []
for line in lines:
    cave.append([int(x) for x in line])
print(dijkstra_grid(cave))
```

```python
# part two
cave = []
for j in range(5):
    for line in lines:
        row = []
        for i in range(5):
            row.extend([int(y) + i + j for y in line])
        row = [z if z < 10 else z - 9 for z in row]
        cave.append(row)
print(dijkstra_grid(cave))
```
