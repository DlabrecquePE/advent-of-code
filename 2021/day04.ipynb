```python
import os
import numpy as np
```

```python
def bingo_check(card, draws):
    card_t = card.transpose()
    for i in range(5):
        if set(card[i]).issubset(draws) or set(card_t[i]).issubset(draws):
            return True
    return False
```

```python
file = 'data/day04.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
bingo_draws = [int(x) for x in lines.pop(0).split(',')]
bingo_cards = []
while lines:
    lines.pop(0)
    bingo_rows = []
    for _ in range(5):
        bingo_rows.append([int(x) for x in lines.pop(0).split()])
    bingo_card = np.array(bingo_rows)
    bingo_cards.append(bingo_card)
```

```python
# part one
nums = []
for i in range(len(bingo_draws)):
    draws = bingo_draws[0:i+1]
    for j, card in enumerate(bingo_cards):
        if bingo_check(card, draws):
            nums = list(set(card.flatten()).difference(draws))
            break
    if nums:
        print(sum(nums) * bingo_draws[i])
        break
```

```python
# part two
w_cards, nums = list(range(len(bingo_cards))), []
for i in range(len(bingo_draws)):
    draws = bingo_draws[0:i+1]
    for j in w_cards:
        if bingo_check(bingo_cards[j], draws):
            if len(w_cards) != 1:
                w_cards.remove(j)
            else:
                nums = set(bingo_cards[w_cards[0]].flatten()).difference(draws)
    if nums:
        print(sum(nums) * bingo_draws[i])
        break
```
