```python
import os
```

```python
file = 'data/day03.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
bit_dict, gamma, epsilon = {}, '', ''
for line in lines:
    for i, bit in enumerate(line):
        bit_dict[i] = bit_dict.get(i, []) + [int(bit)]
for bit in bit_dict:
    bit_sum = sum(bit_dict[bit])
    if bit_sum > len(bit_dict[bit]) / 2:
        gamma += '1'
        epsilon += '0'
        
    else:
        gamma += '0'
        epsilon += '1'
gamma = int(gamma, 2)
epsilon = int(epsilon, 2)
print(gamma * epsilon)
```

```python
# part two
oxy_list = list(range(len(lines)))
co2_list = oxy_list.copy()
for pos in bit_dict:
    oxy_common = [bit_dict[pos][i] for i in oxy_list]
    oxy_check = 1 if sum(oxy_common) >= len(oxy_common) / 2 else 0
    for num in oxy_list.copy():
        if bit_dict[pos][num] != oxy_check:
            oxy_list.remove(num)
    if len(oxy_list) == 1:
        oxy = int(lines[oxy_list[0]], 2)
        break
for pos in bit_dict:
    co2_common = [bit_dict[pos][i] for i in co2_list]
    co2_check = 0 if sum(co2_common) >= len(co2_common) / 2 else 1
    for num in co2_list.copy():
        if bit_dict[pos][num] != co2_check:
            co2_list.remove(num)
    if len(co2_list) == 1:
        co2 = int(lines[co2_list[0]], 2)
        break
print(oxy * co2)
```
