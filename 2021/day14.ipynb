```python
import os
from collections import Counter
```

```python
def polymerize(template, insertion_rules):
    polymer = template[0]
    for i in range(len(template) - 1):
        pair = template[i:i+2]
        polymer += insertion_rules[pair] + pair[1]
    return polymer
```

```python
file = 'data/day14.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
template = lines.pop(0)
insertion_rules = {x: y for x, y in [line.split(' -> ') for line in lines[1:]]}
```

```python
# part one
steps = 10
polymer1 = template
for _ in range(steps):
    polymer1 = polymerize(polymer1, insertion_rules)
element_nums = list(Counter(polymer1).values())
print(max(element_nums) - min(element_nums))
```

```python
twenty_step_dict = {}
for pair in insertion_rules:
    pair_polymer = pair
    for _ in range(20):
        pair_polymer = polymerize(pair_polymer, insertion_rules)
    twenty_step_dict[pair] = Counter(pair_polymer[1:])
```

```python
# part two
polymer2 = template
for _ in range(20):
    polymer2 = polymerize(polymer2, insertion_rules)
total_count = Counter()
for i in range(len(polymer2) - 1):
    pair = polymer2[i:i+2]
    total_count.update(twenty_step_dict[pair])
total_count.update(Counter(template[0]))
element_nums = list(total_count.values())
print(max(element_nums) - min(element_nums))
```
