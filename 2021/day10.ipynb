```python
import os
```

```python
def reduce_string(chunk):
    small_chunks = ['()', '[]', '{}', '<>']
    while True:
        len_check = len(chunk)
        for elem in small_chunks:
            chunk = chunk.replace(elem, '')
        chunk = chunk.rstrip('([{<')
        if len_check == len(chunk):
            return chunk
```

```python
def complete_string(chunk):
    small_chunks = ['()', '[]', '{}', '<>']
    start_end_dict = {'(': ')', '[': ']', '{': '}', '<': '>'}
    while True:
        len_check = len(chunk)
        for elem in small_chunks:
            chunk = chunk.replace(elem, '')
        if len_check == len(chunk):
            break
    return ''.join([start_end_dict[x] for x in chunk])[::-1]
```

```python
file = 'data/day10.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
score = 0
for line in lines:
    reduced = reduce_string(line)
    closers = [')', ']', '}', '>']
    if reduced:
        while True:
            error_char = False
            for char in closers:
                if not reduce_string(reduced[:-1] + char):
                    error_char = reduced[-1]
            if error_char:
                break
            reduced = reduced[:-1]
        match error_char:
            case ')':
                score += 3
            case ']':
                score += 57
            case '}':
                score += 1197
            case '>':
                score += 25137
print(score)
```

```python
# part two
completion_list, score_dict = [], {')': 1, ']': 2, '}': 3, '>': 4}
for line in lines:
    if not reduce_string(line):
        completion_list.append(complete_string(line))
scores = []
for string in completion_list:
    score = 0
    for char in string:
        score = score*5 + score_dict[char]
    scores.append(score)
print(sorted(scores)[len(scores) // 2])
```
