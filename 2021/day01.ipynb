```python
import os
```

```python
file = 'data/day01.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [int(x.rstrip()) for x in fp.readlines()]
```

```python
# part one
prev, count = 0, 0
for x in lines:
    if x > prev:
        count += 1
    prev = x
print(count - 1)
```

```python
# part two
prev, count = 0, 0
for x in range(len(lines) - 2):
    curr = sum(y for y in lines[x:x+3])
    if curr > prev:
        count += 1
    prev = curr
print(count - 1)
```
