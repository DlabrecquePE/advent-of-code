```python
import os
import numpy as np
```

```python
file = 'data/day05.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
vents = []
for line in lines:
    parse = line.replace(' -> ', ',')
    x1, y1, x2, y2 = parse.split(',')
    endpoints = [(int(x1), int(y1)), (int(x2), int(y2))]
    vents.append(endpoints)
x_max = max([x[0][1] for x in vents] + [y[1][1] for y in vents]) + 1
y_max = max([x[0][0] for x in vents] + [y[1][0] for y in vents]) + 1
grid = np.full((x_max,y_max), 0)
for vent in vents:
    x1, y1, x2, y2 = vent[0][0], vent[0][1], vent[1][0], vent[1][1]
    if x1 == x2:
        for i in range(abs(y1 - y2) + 1):
            grid[min([y1, y2]) + i, x1] += 1
    if y1 == y2:
        for j in range(abs(x1 - x2) + 1):
            grid[y1, min([x1, x2]) + j] += 1
print(np.count_nonzero(grid.flatten() > 1))
```

```python
# part two
for vent in vents:
    x1, y1, x2, y2 = vent[0][0], vent[0][1], vent[1][0], vent[1][1]
    if (x1 < x2 and y1 < y2) or (x1 > x2 and y1 > y2):
        for i in range(abs(y1 - y2) + 1):
            grid[min([y1, y2]) + i, min([x1, x2]) + i] += 1
    if (x1 > x2 and y1 < y2) or (x1 < x2 and y1 > y2):
        for j in range(abs(x1 - x2) + 1):
            grid[max([y1, y2]) - j, min([x1, x2]) + j] += 1
print(np.count_nonzero(grid.flatten() > 1))
```
