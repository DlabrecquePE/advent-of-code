```python
import os
import numpy as np
```

```python
file = 'data/day11.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
#part one
steps = 100
grid = np.array([[int(x) for x in line] for line in lines])
row_max, col_max, flashes = len(grid), len(grid[0]), 0
for x in range(steps):
    grid += 1
    reset_list = []
    while True:
        grid_copy = grid.copy()
        for x in range(row_max):
            for y in range(col_max):
                if grid[x, y] > 9:
                    reset_list.append((x, y))
                    i = x - 1 if x > 0 else x
                    j = x + 2 if x < row_max else x + 1
                    k = y - 1 if y > 0 else y
                    l = y + 2 if y < col_max else y + 1
                    grid[i:j, k:l] += 1 
                    for elem in reset_list:
                        row, col = elem
                        grid[row, col] = 0
        if np.array_equal(grid, grid_copy):
            break
    flashes += sum([1 for x in grid.flatten() if x == 0])
print(flashes)
```

```python
# part two
count = 100
while True:
    grid += 1
    count += 1
    reset_list = []
    while True:
        grid_copy = grid.copy()
        for x in range(row_max):
            for y in range(col_max):
                if grid[x, y] > 9:
                    reset_list.append((x, y))
                    i = x - 1 if x > 0 else x
                    j = x + 2 if x < row_max else x + 1
                    k = y - 1 if y > 0 else y
                    l = y + 2 if y < col_max else y + 1
                    grid[i:j, k:l] += 1 
                    for elem in reset_list:
                        row, col = elem
                        grid[row, col] = 0
        if np.array_equal(grid, grid_copy):
            break
    if all(x == 0 for x in grid.flatten()):
        print(count)
        break
```
