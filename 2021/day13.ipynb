```python
import os
import numpy as np
```

```python
file = 'data/day13.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
dot_list = []
while True:
    line = lines.pop(0)
    if line == '':
        break
    dot_list.append(tuple(int(x) for x in line.split(',')))
dot_list = [(y, x) for x, y in dot_list]
fold_list = []
while lines:
    line = lines.pop(0).lstrip('fold along ')
    fold_list.append(tuple(x for x in line.split('=')))
row_max, col_max = max([x[0] for x in dot_list]), max([x[1] for x in dot_list])
grid = np.full((row_max + 1, col_max + 1), '.')
for dot in dot_list:
    grid[dot] = '#'
old_grid = grid.copy()
```

```python
# part one
fold_axis, fold_index = fold_list[0][0], int(fold_list[0][1])
if fold_axis == 'x':
    grid1 = grid[..., :fold_index].copy()
    grid2 = grid[..., :fold_index:-1].copy()
else:
    grid1 = grid[:fold_index,...].copy()
    grid2 = grid[:fold_index:-1, ...].copy()
it = np.nditer(grid2, flags=['multi_index'])
for dot in it:
    if dot == '#':
        grid1[it.multi_index] = '#'
grid = grid1
dots = list(grid.flatten())
print(dots.count('#'))
```

```python
# part two
grid = old_grid
for fold in fold_list:
    fold_axis, fold_index = fold[0], int(fold[1])
    if fold_axis == 'x':
        grid1 = grid[..., :fold_index].copy()
        grid2 = grid[..., :fold_index:-1].copy()
    else:
        grid1 = grid[:fold_index,...].copy()
        grid2 = grid[:fold_index:-1, ...].copy()
    it = np.nditer(grid2, flags=['multi_index'])
    for dot in it:
        if dot == '#':
            grid1[it.multi_index] = '#'
    grid = grid1
for row in grid:
    print(''.join(row))
```
