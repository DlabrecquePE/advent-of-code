```python
import os
```

```python
file = 'data/day02.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
horizontal, depth = 0, 0
for line in lines:
    if line.startswith('forward'):
        horizontal += int(line.strip('forward '))
    if line.startswith('up'):
        depth -= int(line.strip('up '))
    if line.startswith('down'):
        depth += int(line.strip('down '))
print(horizontal * depth)
```

```python
# part two
horizontal, depth, aim = 0, 0, 0
for line in lines:
    if line.startswith('forward'):
        forward = int(line.strip('forward '))
        horizontal += forward
        depth += forward * aim
    if line.startswith('up'):
        aim -= int(line.strip('up '))
    if line.startswith('down'):
        aim += int(line.strip('down '))
print(horizontal * depth)
```
