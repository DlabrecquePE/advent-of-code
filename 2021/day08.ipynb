```python
import os
```

```python
file = 'data/day08.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
outputs = []
for line in lines:
    outputs.append(line.split(' | ')[1])
digits_1478 = []
for output in outputs:
    digit_list = output.split()
    for digit in digit_list:
        if len(digit) in [2, 3, 4, 7]:
            digits_1478.append(digit)
print(len(digits_1478))
```

```python
# part two
signals, outputs, solutions = [], [], []
for line in lines:
    signal, output = line.split(' | ')
    signals.append(signal)
    outputs.append(output)
output_list = [output.split() for output in outputs]
for i, signal in enumerate(signals):
    solution_dict = {}
    digit_list = signal.split()
    for digit in digit_list:
        if len(digit) == 2:
            solution_dict[1] = digit
        if len(digit) == 3:
            solution_dict[7] = digit
        if len(digit) == 4:
            solution_dict[4] = digit
        if len(digit) == 7:
            solution_dict[8] = digit
    for digit in digit_list:
        if len(digit) == 6:
            letter_set = set(x for x in digit)
            if not letter_set.issuperset(solution_dict[1]):
                solution_dict[6] = digit
    for digit in digit_list:
        if len(digit) == 5:
            letter_set = set(x for x in digit)
            if letter_set.issubset(solution_dict[6]):
                solution_dict[5] = digit
            elif letter_set.issuperset(solution_dict[1]):
                solution_dict[3] = digit
            else:
                solution_dict[2] = digit
    missing_segment = set(solution_dict[6]).difference(solution_dict[5]).pop()
    for digit in digit_list:
        if len(digit) == 6:
            if missing_segment not in digit:
                solution_dict[9] = digit
    digit_set, solutions_set = set(digit_list), set(solution_dict.values())
    solution_dict[0] = list(digit_set.difference(solutions_set))[0]
    solution_dict = {value: key for key, value in solution_dict.items()}
    output_values = []
    for out in output_list[i]:
        for key in solution_dict:
            if set(key) == set(out):
                output_values.append(str(solution_dict[key]))
    solutions.append(int(''.join(output_values)))
print(sum(solutions))
    
```
