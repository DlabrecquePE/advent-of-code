```python
import os
```

```python
file = 'data/day07.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    line = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
h_pos_list = [int(x) for x in line[0].split(',')]
h_max, fuel_reqs = max(h_pos_list), []
for x in range(h_max + 1):
    fuel_reqs.append(sum([abs(x - y) for y in h_pos_list]))
print(min(fuel_reqs))
```

```python
# part two
fuel_reqs = []
for x in range(h_max + 1):
    fuel_reqs.append(sum([(abs(x-y)**2 + abs(x-y)) // 2 for y in h_pos_list]))
print(min(fuel_reqs))
```
