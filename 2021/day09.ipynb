```python
import os
import numpy as np
```

```python
file = 'data/day09.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
rows, risk_levels = [], []
for line in lines:
    rows.append([int(x) for x in line])
grid = np.array(rows)
row_max, col_max = len(grid), len(grid[0])
for i in range(row_max):
    for j in range(col_max):
        adjacent_points = []
        grid_point = grid[i, j]
        if i > 0:
            adjacent_points.append(grid[i-1, j])
        if i < row_max-1:
            adjacent_points.append(grid[i+1, j])
        if j > 0:
            adjacent_points.append(grid[i, j-1])
        if j < col_max-1:
            adjacent_points.append(grid[i, j+1])
        if grid_point < min(adjacent_points):
             risk_levels.append(1 + grid_point)
print(sum(risk_levels))
```

```python
# part two
checked, basin_dict, basin_count = set(), {}, 0
for i in range(row_max):
    for j in range(col_max):
        if (i, j) not in checked and grid[i, j] != 9:
            new_points, basin_list = {(i, j)}, []
            while new_points:
                basin_list.extend(new_points)
                checked.update(new_points)
                new_points, old_points = set(), new_points
                for point in old_points:
                    k, l = point
                    if (k > 0 and grid[k-1, l] != 9 and
                        (k-1, l) not in checked):
                        new_points.add((k-1, l))
                    if (k < row_max-1 and grid[k+1, l] != 9 and
                        (k+1, l) not in checked):
                        new_points.add((k+1, l))
                    if (l > 0 and grid[k, l-1] != 9 and 
                        (k, l-1) not in checked):
                        new_points.add((k, l-1))
                    if (l < col_max-1 and grid[k, l+1] != 9 and 
                        (k, l+1) not in checked):
                        new_points.add((k, l+1))
            basin_dict[basin_count] = basin_list
            basin_count += 1
basin_sizes = sorted([len(basin_dict[x]) for x in basin_dict])
a, b, c = basin_sizes[-3:]
print(a * b * c)
```
