```python
import os
```

```python
file = 'data/day12.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
path_dict = {}
for line in lines:
    from_point, to_point = line.split('-')
    if from_point != 'end' and to_point != 'start':
        path_dict[from_point] = path_dict.get(from_point, []) + [to_point]
    if to_point != 'end' and from_point != 'start':
        path_dict[to_point] = path_dict.get(to_point, []) + [from_point]
path_dict = {x: set(y) for x, y in path_dict.items()}
for key, values in path_dict.copy().items():
    for value in values.copy():
        if key == value:
            path_dict[key].remove(value)
```

```python
# part one
start_node, end_node = 'start', 'end'
path_list, wip_paths = [], [[start_node]]
while wip_paths:
    for path in wip_paths.copy():
        current_node = path[-1]
        if current_node == 'end':
            path_list.append(path)
            wip_paths.remove(path)
            continue
        for node in path_dict[current_node]:
            if node not in path or node.isupper():
                wip_paths.append(path + [node])
        wip_paths.remove(path)
print(len(path_list))
```

```python
# part two
start_node, end_node = 'start', 'end'
path_list, wip_paths = [], [[True, start_node]]
while wip_paths:
    for path in wip_paths.copy():
        current_node = path[-1]
        if current_node == 'end':
            path_list.append(path)
            wip_paths.remove(path)
            continue
        for node in path_dict[current_node]:
            if path[0] and (node in path) and node.islower():
                wip_paths.append([False] + path[1:] + [node])
            if node not in path or node.isupper():
                wip_paths.append(path + [node])
        wip_paths.remove(path)
print(len(path_list))
```
