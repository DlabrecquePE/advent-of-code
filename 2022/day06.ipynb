```python
import os
```

```python
file = 'data/day06.txt'
path = os.path.join(os.getcwd(), file)
fp = open(path, 'r')
text = fp.read()
```

```python
# part one
index, packet_len = 0, 4
while True:
    packet = [x for x in text[index:index + packet_len]]
    if len(packet) == len(set(packet)):
        print(index + packet_len)
        break
    index += 1
```

```python
# part two
index, packet_len = 0, 14
while True:
    packet = [x for x in text[index:index + packet_len]]
    if len(packet) == len(set(packet)):
        print(index + packet_len)
        break
    index += 1
```
