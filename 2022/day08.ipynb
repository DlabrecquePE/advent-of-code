```python
import os
import numpy as np
```

```python
def vt_count(tree_string):
    tallest, vis_trees = -1, []
    for i, tree in enumerate(tree_string):
        height = int(tree)
        if height > tallest:
            tallest = height
            vis_trees.append(i)
    return vis_trees


def s_count(tree_string):
    height, visibility = tree_string[0], 0
    for tree in tree_string[1:]:
        visibility += 1
        if height <= tree:
            break
    return visibility
```

```python
file = 'data/day08.txt'
path = os.path.join(os.getcwd(), file)
fp = open(path, 'r')
rows = [x.strip('\n') for x in fp.readlines()]
```

```python
# part one
s_len, vt_coords = len(rows), set()
cols = [''] * s_len
for y, row in enumerate(rows[::-1]):
    vt_coords.update({(x, y) for x in vt_count(row)})
    vt_coords.update({(s_len - x - 1, y) for x in vt_count(row[::-1])[::-1]})
    for j, elem in enumerate(row):
        cols[j] += elem
for x, col in enumerate(cols):
    vt_coords.update({(x, y) for y in vt_count(col)})
    vt_coords.update({(x, s_len - y - 1) for y in vt_count(col[::-1])})
print(len(vt_coords))    
```

```python
# part two
grid = np.array([[int(elem) for elem in row] for row in rows[::-1]])
grid_t, high_score = grid.transpose(), 0
for x in range(s_len):
    for y in range(s_len):
        score = s_count(grid[y][x:]) * s_count(grid[y][x::-1])
        score *= s_count(grid_t[x][y:]) * s_count(grid_t[x][y::-1])
        if score > high_score:
            high_score = score
print(high_score)
```
