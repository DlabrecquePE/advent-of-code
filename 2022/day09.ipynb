```python
import os
```

```python
def knot_pos(h_pos, t_pos):
    ht_diff = (h_pos[0] - t_pos[0], h_pos[1] - t_pos[1])
    if any(x in ht_diff for x in [-2, 2]):
        t_mov = tuple(x // 2 if abs(x) == 2 else x for x in ht_diff)
    else:
        t_mov = (0, 0)
    t_pos = tuple(map(sum, zip(t_pos, t_mov)))
    return t_pos
```

```python
file = 'data/day09.txt'
path = os.path.join(os.getcwd(), file)
fp = open(path, 'r')
rows = [x.strip('\n') for x in fp.readlines()]
```

```python
# part one
t_pos, h_pos, inst = (0, 0), (0, 0), ''
t_pos_set = {t_pos}
inst_dict = {'U': (0, 1),'D': (0, -1),'L': (-1, 0),'R': (1, 0)}
for row in rows:
    direct, mult = row.split()
    inst += direct * int(mult)
for direct in inst:
    # print(direct,'t',t_pos,'h',h_pos)
    h_pos = tuple(map(sum, zip(h_pos, inst_dict[direct])))
    t_pos = knot_pos(h_pos, t_pos)
    t_pos_set.add(t_pos)
print(len(t_pos_set))
```

```python
# part two
h_pos = (0, 0)
knots, t_pos_set = [h_pos,] * 10, {h_pos}
for direct in inst:
    knots[0] = tuple(map(sum, zip(knots[0], inst_dict[direct])))
    for i, knot in enumerate(knots.copy()[1:]):
        knots[i + 1] = knot_pos(knots[i], knots[i + 1])
    t_pos_set.add(knots[9])
print(len(t_pos_set))
```
