```python
import os
import numpy as np
```

```python
def find_path(pos, grid):
    row, col = pos
    c_max, r_max = len(grid[0]), len(grid)
    pos_elev, work = grid[row][col], []
    if row + 1 < r_max:
        d_elev = grid[row + 1][col]
        if d_elev >= pos_elev - 1:
            work.append((row + 1, col))
    if row - 1 >= 0:
        u_elev = grid[row - 1][col]
        if u_elev >= pos_elev - 1:
            work.append((row - 1, col))
    if col + 1 < c_max:
        r_elev = grid[row][col + 1]
        if r_elev >= pos_elev - 1:
            work.append((row, col + 1))
    if col - 1 >= 0:
        l_elev = grid[row][col - 1]
        if l_elev >= pos_elev - 1:
            work.append((row, col - 1))
    return work
```

```python
file = 'data/day12.txt'
path = os.path.join(os.getcwd(), file)
fp = open(path, 'r')
rows = [x.strip('\n') for x in fp.readlines()]
```

```python
# part one
for i, row in enumerate(rows):
    start_col, end_col = row.find('S'), row.find('E')
    if start_col != -1:
        start_pos = (i, start_col)
    if end_col != -1:
        end_pos = (i, end_col)
grid = np.array([[ord(x) - 97 for x in row] for row in rows])
grid[end_pos[0]][end_pos[1]] = 25
grid[start_pos[0]][start_pos[1]] = 0
pos, path_set, path_len = end_pos, {end_pos}, 1
work = find_path(pos, grid)
while True:
    if start_pos in work:
        break
    new_work = []
    for step in work:
        if not step in path_set:
            path_set.add(step)
            new_work.append(step)
    work = []
    for step in new_work:
        work.extend(find_path(step, grid))
    path_len += 1
print(path_len)
```

```python
# part two
a_pos_set = set()
for i, row in enumerate(rows):
    for j, col in enumerate(row):
        if grid[i][j] == 0:
            a_pos_set.add((i, j))
pos, path_set, path_len = end_pos, {end_pos}, 1
work = find_path(pos, grid)
while True:
    if a_pos_set.intersection(set(work)):
        break
    new_work = []
    for step in work:
        if not step in path_set:
            path_set.add(step)
            new_work.append(step)
    work = []
    for step in new_work:
        work.extend(find_path(step, grid))
    path_len += 1
print(path_len)
```
