```python
import os
```

```python
file = 'data/day02.txt'
path = os.path.join(os.getcwd(), file)
fp = open(path, 'r')
lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
scores = []
score_converter = {'X': 1, 'Y': 2, 'Z': 3} # score for rock paper scissors
for line in lines:
    score = score_converter[line[2]]
    if line in ['A Y', 'B Z', 'C X']: # win
        scores.append(6 + score)
    if line in ['A Z', 'B X', 'C Y']: # lose
        scores.append(0 + score)
    if line in ['A X', 'B Y', 'C Z']: # draw
        scores.append(3 + score)
print(sum(scores))
```

```python
# part two
scores = []
score_converter = {'X': 0, 'Y': 3, 'Z': 6} # score for loss draw win
for line in lines:
    score = score_converter[line[2]]
    if line in ['A Y', 'B X', 'C Z']: # rock
        scores.append(1 + score)
    if line in ['A Z', 'B Y', 'C X']: # paper
        scores.append(2 + score)
    if line in ['A X', 'B Z', 'C Y']: # scissors
        scores.append(3 + score)
print(sum(scores))
```
