```python
import os
```

```python
file = 'data/day05.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip('\n') for x in fp.readlines()]
```

```python
# populate initial stack configuration
stacks = {key: [] for key in range(1,10)}
for _ in range(8):
    line = lines.pop(0)
    for stack in range(9):
        start = stack * 4
        end = start + 3
        if line[start:end] != '   ':
            stacks[stack + 1].append(line[stack*4 + 1])
del(lines[0:2])
```

```python
# part one
stack_dict = stacks.copy()
for line in lines:
    instruct = line.split()
    quant, source, dest = int(instruct[1]), int(instruct[3]), int(instruct[5])
    stack_dict[dest] = stack_dict[source][quant - 1::-1] + stack_dict[dest]
    stack_dict[source] = stack_dict[source][quant:]
top_string = ''
for stack in stack_dict:
    if stack_dict[stack]:
        top_string += stack_dict[stack][0]
print(top_string)
```

```python
# part two
stack_dict = stacks.copy()
for line in lines:
    instruct = line.split()
    quant, source, dest = int(instruct[1]), int(instruct[3]), int(instruct[5])
    stack_dict[dest] = stack_dict[source][:quant:] + stack_dict[dest]
    stack_dict[source] = stack_dict[source][quant:]
top_string = ''
for stack in stack_dict:
    if stack_dict[stack]:
        top_string += stack_dict[stack][0]
print(top_string)
```
