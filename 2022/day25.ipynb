```python
import os
```

```python
def snafu_to_decimal(snafu_str):
    digit_dict = {'0': 0, '1': 1, '2': 2, '-': -1, '=': -2}
    res = 0
    for power, digit in enumerate(snafu_str[::-1]):
        res += (5 ** power) * digit_dict[digit]
    return res
```

```python
def decimal_to_snafu(number):
    base_5 = ''
    while number:
        base_5 = str(number % 5) + base_5
        number = number // 5
    digit_converter = {0: '0', 1: '1', 2: '2', 3: '=', 4: '-', 5: '0'}
    snafu_string, carry = '', 0
    for digit in base_5[::-1]:
        snafu_digit = int(digit) + carry
        if snafu_digit in [0, 1, 2]:
            carry = 0
        else:
            carry = 1
        snafu_string = digit_converter[snafu_digit] + snafu_string
    return snafu_string
```

```python
file = 'data/day25.txt'
path = os.path.join(os.getcwd(), file)
fp = open(path, 'r')
decimal_total = 0
for line in fp.readlines():
    decimal_total += snafu_to_decimal(line.rstrip())
print(decimal_to_snafu(decimal_total))
```
