```python
import os
import numpy as np
```

```python
file = 'data/day14.txt'
path = os.path.join(os.getcwd(), file)
fp = open(path, 'r')
rows = [x.strip('\n') for x in fp.readlines()]
```

```python
# Initialize grid
paths = []
for row in rows:
    path = [x for x in row.split(' -> ')]
    path = [(int(x), int(y)) for x, y in [z.split(',') for z in path]]
    paths.append(path)
col_coords = [[x[0] for x in path] for path in paths]
col_coords = [y for x in col_coords for y in x]
row_coords = [[x[1] for x in path] for path in paths]
row_coords = [y for x in row_coords for y in x]
col_min, col_max = min(col_coords), max(col_coords) + 1
row_min, row_max = 0, max(row_coords) + 1
grid_width, grid_height = col_max - col_min, row_max - row_min
grid = np.full((grid_height, grid_width), '.')
```

```python
# fill grid
for path in paths:
    old_coord = path.pop(0)
    for new_coord in path:
        coord_diff = (old_coord[0] - new_coord[0], old_coord[1] - new_coord[1])
        row, col = old_coord[1], old_coord[0] - col_min
        grid[row][col] = '#'
        if coord_diff[0] == 0:
            if coord_diff[1] > 0:
                for i in range(coord_diff[1] + 1):
                    grid[row - i][col] = '#'
            else:
                for i in range(abs(coord_diff[1]) + 1):
                    grid[row + i][col] = '#'
        elif coord_diff[0] > 0:
            for i in range(coord_diff[0] + 1):
                grid[row][col - i] = '#'
        else:
            for i in range(abs(coord_diff[0]) + 1):
                grid[row][col + i] = '#'
        old_coord = new_coord
```

```python
# part one
grid_one = grid.copy()
sand_origin = (0, 500 - col_min)
sand_pos = sand_origin
sand_units = 0
while True:
    row, col = sand_pos[0], sand_pos[1]
    if row + 1 < grid_height:
        if grid_one[row + 1][col] == '.':
            sand_pos = (row + 1, col)
            continue
    else:
        break
    if col - 1 > 0:
        if grid_one[row + 1][col - 1] == '.':
            sand_pos = (row + 1, col - 1)
            continue
    else:
        break
    if col + 1 < grid_width:
        if grid_one[row + 1][col + 1] == '.':
            sand_pos = (row + 1, col + 1)
            continue
    else:
        break
    grid_one[row][col] = 'o'
    sand_units += 1
    sand_pos = sand_origin
print(sand_units)
```

```python
# part two
grid_two = grid.copy()
grid_two = np.append(grid_two, np.full((1, len(grid_two[0])), '.'), axis = 0)
grid_two = np.append(grid_two, np.full((1, len(grid_two[0])), '#'), axis = 0)
grid_extend = np.full((len(grid_two), 1), '.')
grid_extend[-1] = '#'
sand_origin = (0, 500 - col_min)
sand_pos = sand_origin
sand_units = 0
while True:
    row, col = sand_pos[0], sand_pos[1]
    if grid_two[row + 1][col] == '.':
        sand_pos = (row + 1, col)
        continue
    if col - 1 > 0:
        if grid_two[row + 1][col - 1] == '.':
            sand_pos = (row + 1, col - 1)
            continue
    else:
        grid_two = np.append(grid_extend, grid_two, axis = 1)
        sand_origin = (0, sand_origin[1] + 1)
        sand_pos = (row, col + 1)
        grid_width += 1
        continue
    if col + 1 < grid_width:
        if grid_two[row + 1][col + 1] == '.':
            sand_pos = (row + 1, col + 1)
            continue
    else:
        grid_two = np.append(grid_two, grid_extend, axis = 1)
        grid_width += 1
        continue
    grid_two[row][col] = 'o'
    sand_units += 1
    if (row, col) == sand_origin:
        break
    sand_pos = sand_origin
print(sand_units)
```
