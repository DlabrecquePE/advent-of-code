```python
import os
```

```python
file = 'data/day01.txt'
path = os.path.join(os.getcwd(), file)
fp = open(path, 'r')
lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
elf_packs, pack = [], 0
for line in lines:
    if line:
        pack += int(line)
    else:
        elf_packs.append(pack)
        pack = 0
print(max(elf_packs))
```

```python
# part two
top_three = []
for _ in range(3):
    top = max(elf_packs)
    top_three.append(top)
    elf_packs.remove(top)
print(sum(top_three))
```
