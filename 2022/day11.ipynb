```python
import os
import copy
```

```python
file = 'data/day11.txt'
path = os.path.join(os.getcwd(), file)
fp = open(path, 'r')
rows = [x.strip('\n') for x in fp.readlines()]
```

```python
# part one
monkey_input, activity_dict, m_div = {}, {}, []
while rows: # parsing input
    monkey = int(rows.pop(0).strip().split()[1][0])
    inv_string = rows.pop(0).strip().replace('Starting items: ', '')
    inv = [int(x) for x in inv_string.split(', ')]
    op_string = rows.pop(0).strip().replace('Operation: new = old ', '')
    op, num = [x for x in op_string.split()]
    test = int(rows.pop(0).strip().replace('Test: divisible by ', ''))
    t_num = int(rows.pop(0).strip().replace('If true: throw to monkey ', ''))
    f_num = int(rows.pop(0).strip().replace('If false: throw to monkey ', ''))
    if rows:
        rows.pop(0)
    monkey_input[monkey] = [inv, op, num, t_num, f_num]
    activity_dict[monkey] = 0
    m_div.append(test)

monkeys = copy.deepcopy(monkey_input)
activity = activity_dict.copy()
for _ in range(20): # processing rounds
    for m in monkeys:
        inv, op, num, t_num, f_num = monkeys[m]
        for item in inv.copy():
            number = item if num == 'old' else int(num)
            if op == '+':
                worry = (number + item) // 3
            else:
                worry = (number * item) // 3
            if not worry % m_div[m]:
                monkeys[t_num][0].append(worry)
            else:
                monkeys[f_num][0].append(worry)
            monkeys[m][0].remove(item)
            activity[m] += 1
top_two = sorted(list(activity.values()))[-2:]
print(top_two[0] * top_two[1])
```

```python
# part two 
monkeys = copy.deepcopy(monkey_input)
activity, m_size = activity_dict.copy(), len(monkeys)
for m in monkeys: # init for modular arithmetic
    inv, num = monkeys[m][0], monkeys[m][2]
    monkeys[m][0] = [[x] * m_size for x in inv]
    monkeys[m][2] = [num for _ in range(m_size)]

for _ in range(10000):
    for m in monkeys:
        inv, op, nums, t_num, f_num = monkeys[m]
        for item in inv.copy():
            ns = [] # using list of residues to keep worry manageable
            for i, num in enumerate(nums):
                n = item[i] if num == 'old' else int(num)
                ns.append(n % m_div[i])
            if op == '+':
                worry = [ns[i] + (item[i] % d) for i, d in enumerate(m_div)]
            else:
                worry = [ns[i] * (item[i] % d) for i, d in enumerate(m_div)]
            if not worry[m] % m_div[m]:
                monkeys[t_num][0].append(worry)
            else:
                monkeys[f_num][0].append(worry)
            monkeys[m][0].remove(item)
            activity[m] += 1
top_two = sorted(list(activity.values()))[-2:]
print(top_two[0] * top_two[1])
```
