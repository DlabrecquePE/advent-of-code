```python
import os
```

```python
file = 'data/day10.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip('\n') for x in fp.readlines()]
```

```python
# part one
cycle, x, op_cycle, op = 1, 1, 1, lines.copy()
signal_list, value = [], 0
out = ''
while op:
    if cycle == op_cycle:
        instruction = op.pop(0)
        x += value
        if instruction.startswith('addx'):
            op_cycle += 2
            value = int(instruction.split()[1])
        else:
            op_cycle += 1
            value = 0
    if cycle in [20, 60, 100, 140, 180, 220]:
        signal_list.append(x * cycle)
    if (cycle - 1) % 40 in [x - 1, x, x + 1]:
        out += '#'
    else:
        out += '.'
    cycle += 1
print(sum(signal_list))
```

```python
# part two
n = 40
screen_lines = [out[i:i + n] for i in range(0, len(out), n)]
for line in screen_lines:
    print(line)
```
