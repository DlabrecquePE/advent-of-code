```python
import os
```

```python
file = 'data/day03.txt'
path = os.path.join(os.getcwd(), file)
fp = open(path, 'r')
lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
priorities = []
for line in lines:
    index = len(line) // 2
    first_half = {elem for elem in line[:index]}
    second_half = {elem for elem in line[index:]}
    item_type = ''.join(first_half.intersection(second_half))
    if item_type.isupper():
        priorities.append(ord(item_type) - 38)
    else:
        priorities.append(ord(item_type) - 96)
print(sum(priorities))
```

```python
# part two
priorities = []
while lines:
    item_type = ''.join(set(lines.pop()) & set(lines.pop()) & set(lines.pop()))
    if item_type.isupper():
        priorities.append(ord(item_type) - 38)
    else:
        priorities.append(ord(item_type) - 96)
print(sum(priorities))
```
