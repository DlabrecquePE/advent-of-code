```python
import os
```

```python
file = 'data/day07.txt'
path = os.path.join(os.getcwd(), file)
fp = open(path, 'r')
lines = [x.strip('\n') for x in fp.readlines()]
```

```python
# populate directory tree
index, cwd = 0, ('/',)
dir_tree, inst_len = {cwd: [0]}, len(lines)
while index < inst_len:
    command = lines[index]
    arg = list(command.split())
    if command.startswith('$ cd'):
        if command.endswith(' ..'):
            cwd = cwd[:-1]
        elif command.endswith(' /'):
            cwd = ('/',)
        else:
            cwd = cwd + (arg[2],)
    elif command.startswith('$ ls'):
        pass
    elif command.startswith('dir '):
        new_dir = cwd + (arg[1],)
        dir_tree[cwd].append(arg[1])
        dir_tree[new_dir] = dir_tree.get(new_dir, [0])
    else:
        dir_tree[cwd][0] += int(arg[0])
    index += 1
```

```python
# part one
tree_copy, size_dict = dir_tree.copy(), {}
while tree_copy:
    cull_list = [elem for elem in tree_copy if len(tree_copy[elem]) == 1]
    for dir in cull_list:
        size, parent = tree_copy[dir][0], dir[:-1]
        if parent:
            tree_copy[parent][0] += size
            tree_copy[parent].remove(dir[-1])
        del tree_copy[dir]
        size_dict[dir] = size
print(sum([elem for elem in size_dict.values() if elem < 100000]))
```

```python
# part two
free_space = 70000000 - size_dict[('/',)]
free_up = 30000000 - free_space
dir_spaces = sorted(list(size_dict.values()))
for i, x in enumerate(dir_spaces):
    if x > free_up:
        print(dir_spaces[i])
        break
```
