```python
import os
```

```python
def packet_comp(left, right):
    l, r = left.pop(0), right.pop(0)
    if l == ']' and r != ']':
        return True
    if r == ']' and l != ']':
        return False
    if l == '[' and r == '[':
        return packet_comp(left, right)
    if l == '[' and r != '[':
        right = ['[', r, ']'] + right
        left = [l] + left
        return packet_comp(left, right)
    if r == '[' and l != '[':
        left = ['[', l, ']'] + left
        right = [r] + right
        return packet_comp(left, right)
    if l == ']' and r == ']':
        return packet_comp(left, right)
    l, r = int(l), int(r)
    if l < r:
        return True
    if l > r:
        return False
    if l == r:
        return packet_comp(left, right)
```

```python
file = 'data/day13.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip('\n') for x in fp.readlines()]
```

```python
# part one
packets = []
for row in lines:
    if row != '':
        new_row = row.replace('[', '[,').replace(']', ',],')
        packets.append([x for x in new_row.split(',') if x != ''])
index, index_list = 1, []
while packets:
    left = packets.pop(0)
    right = packets.pop(0)
    if packet_comp(left, right):
        index_list.append(index)
    index += 1 
print(sum(index_list))
```

```python
# part two
packets = []
div1, div2 = '[[2]]', '[[6]]'
lines.extend([div1, div2])
for row in lines:
    if row != '':
        new_row = row.replace('[', '[,').replace(']', ',],')
        packets.append([x for x in new_row.split(',') if x != ''])
index, index_list = 1, []
while True:
    swap_test = packets.copy()
    for i in range(len(packets) - 1):
        left, right = packets[i], packets[i + 1]
        if packet_comp(right.copy(), left.copy()):
            packets[i], packets[i + 1] = right, left
    if swap_test == packets:
        break
div1_index = packets.index(['[', '[', '2', ']', ']']) + 1
div2_index = packets.index(['[', '[', '6', ']', ']']) + 1
print(div1_index * div2_index)
```
