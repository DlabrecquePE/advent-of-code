```python
import os
```

```python
file = 'data/day04.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
pairs = 0
for line in lines:
    sections = [list(int(x) for x in y.split('-')) for y in line.split(',')]
    a, b, c, d = *sections[0], *sections[1]
    if (a >= c and b <= d) or (a <= c and b >= d):
        pairs += 1
print(pairs)
```

```python
# part two
pairs = 0
for line in lines:
    sections = [list(int(x) for x in y.split('-')) for y in line.split(',')]
    a, b, c, d = *sections[0], *sections[1]
    if (a >= c and a <= d) or (c >= a and c <= b):
        pairs += 1
print(pairs)
```
