```python
import os
```

```python
file = 'data/day01.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    line = [int(x) for x in fp.read() if x.strip()]
```

```python
# part 1
line = line + [line[0]]
captcha_sum = 0
for i, x in enumerate(line[:-1]):
    if x == line[i + 1]:
        captcha_sum += x
print(captcha_sum)
```

```python
# part 2
captcha_sum, line_len = 0, len(line[:-1])
line1, line2 = line[:line_len // 2], line[line_len // 2:-1]
while line1:
    num = line1.pop()
    if num == line2.pop():
        captcha_sum += (2 * num)
print(captcha_sum)
```
