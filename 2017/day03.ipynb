```python
# part two
line = 26
line_root = ceil(line ** 0.5)
grid_size = line_root if line_root % 2 else line_root + 1
grid_vol = grid_size ** 2
```

```python
# part two
line = 8
line_root = ceil(line ** 0.5)
grid_size = line_root if line_root % 2 else line_root + 1
min_square = line_root ** 2 if line_root % 2 else (line_root - 1) ** 2
max_square = (min_square + 2) ** 2
perimeter_len = max_square - min_square
side_len = perimeter_len // 4
residual = line - min_square + 1
side_num = side_len // residual
center = side_len // 2 + 1
match side_num:
    case 0:
        target = (min_square + 1, residual)
    case 1:
        target = (min_square + 1 - (residual - side_len), 0)
    case 2:
        target = (0, residual - 2 * side_len)
    case 3:
        target = (0 + residual - 3 * side_len, min_square + 1)
manhattan = abs(target[0] - center) + abs(target[1] - center)
print(manhattan)
```
