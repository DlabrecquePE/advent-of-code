```python
import os
from itertools import permutations
```

```python
file = 'data/day02.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
lines =[[int(x) for x in line.split('\t')] for line in lines]
```

```python
# part one
checksum = 0
for line in lines:
    checksum += (max(line) - min(line))
print(checksum)
```

```python
# part two
checksum = 0
for line in lines:
    for x, y in permutations(line, 2):
        if x % y == 0:
            checksum += x // y
print(checksum)
```
