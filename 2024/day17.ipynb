```python
import os
```

```python
file = 'data/day17.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip() for x in fp.readlines()]
```

```python
reg_a = int(lines.pop(0).strip('Register A: '))
reg_b = int(lines.pop(0).strip('Register B: '))
reg_c = int(lines.pop(0).strip('Register C: '))
program = list(map(int, lines.pop().strip('Program: ').split(',')))
```

```python
def compute(program, reg_a, reg_b, reg_c):
    inst_pointer, output = 0, []
    program = {k: v for k, v in enumerate(program)}
    while True:
        opcode = program[inst_pointer]
        operand = program[inst_pointer + 1]
        match operand:
            case 0 | 1 | 2 | 3:
                combo = operand
            case 4:
                combo = reg_a
            case 5:
                combo = reg_b
            case 6:
                combo = reg_c
        match opcode:
            case 0: # adv
                reg_a = reg_a // 2 ** combo
            case 1: # bxl
                reg_b = reg_b ^ operand
            case 2: # bst
                reg_b = combo % 8
            case 3: # jnz
                inst_pointer = operand if reg_a else inst_pointer + 2
            case 4: # bxc
                reg_b = reg_b ^ reg_c
            case 5: # out
                output.append(combo % 8)
            case 6: # bdv
                reg_b = reg_a // 2 ** combo
            case 7: # cdv
                reg_c = reg_a // 2 ** combo
        inst_pointer = inst_pointer + 2 if opcode != 3 else inst_pointer
        if len(program) <= inst_pointer:
            return output
```

```python
# part one
print(','.join(str(x) for x in compute(program, reg_a, reg_b, reg_c)))    
print(len(program))
```
