```python
import os
from collections import defaultdict
```

```python
file = 'data/day05.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
order_dict = defaultdict(set)
update_lists = []
while True:
    line = lines.pop()
    if line == '':
        break
    update_lists.append([int(x) for x in line.split(',')])
while lines:
    line = lines.pop()
    high, low = line.split('|')
    order_dict[int(high)].add(int(low))
```

```python
# part one
correct_list, incorrect_list = [], []
for update_list in update_lists:
    updates = update_list.copy()
    while updates:
        test = updates.pop()
        if order_dict[test].intersection(updates):
            incorrect_list.append(update_list)
            break
    else:
        correct_list.append(update_list[len(update_list) // 2])
print(sum(correct_list))
```

```python
# part two
corrected_list = []
for wrong in incorrect_list:
    corrected = []
    while wrong:
        head = wrong[0]
        rule = order_dict[head]
        tail = wrong[1:]
        if not rule.intersection(tail):
            corrected.append(head)
            wrong = tail
        else:
            wrong = tail + [head]
    corrected_list.append(corrected[len(corrected) // 2])
print(sum(corrected_list))
```
