```python
import os
```

```python
file = 'data/day15.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip() for x in fp.readlines()]
```

```python
def move(pos, dir):
    row, col = pos
    move_dict = {'<': (row, col - 1), '>': (row, col + 1),
                 '^': (row - 1, col), 'v': (row + 1, col)}
    return move_dict[dir]
```

```python
def print_grid(grid, max_r, max_c):
    for row in range(max_r):
        print(''.join([grid[(row, col)] for col in range(max_c)]))
```

```python
def get_part2_coords(grid, dir):
    boxes = {k for k, v in grid.items() if v == '['}
    boxes.update([(box[0], box[1] + 1) for box in boxes])
    pos = list(grid.keys())[list(grid.values()).index('@')]
    move_list = []
    test_list = [pos]
    while True:
        new_boxes = []
        for new_pos in test_list:
            row, col = new_pos
            if dir == '^':
                u, ul, ur = (row - 1, col), (row - 1, col - 1), (row - 1, col + 1)
                if (u_value := grid[u]) in ['[', ']']:
                    new_val = [ul, u] if u_value == ']' else [u, ur]
                    for val in new_val:
                        if val not in new_boxes:
                            new_boxes.append(val)
            if dir == 'v':
                d, dl, dr = (row + 1, col), (row + 1, col - 1), (row + 1, col + 1)
                if (d_value := grid[d]) in ['[', ']']:
                    new_val = [dl, d] if d_value == ']' else [d, dr]
                    for val in new_val:
                        if val not in new_boxes:
                            new_boxes.append(val)
        if new_boxes:
            test_list = new_boxes
            move_list.extend(new_boxes)
        else:
            break
    return move_list
```

```python
grid_master, grid, movements = [], {}, ''
while lines[0] != '':
    grid_master.append(lines.pop(0))
for row, line in enumerate(grid_master):
    for col, value in enumerate(line):
        if value == '@':
            start_pos = (row, col)
        grid[(row, col)] = value
lines.pop(0)
while lines:
    movements += lines.pop(0)
max_rows, max_cols = len(grid_master), len(grid_master[0])
```

```python
pos = start_pos
for dir in movements:
    row, col = pos
    rel_pos, work_list = pos, []
    while rel_pos in grid and grid[rel_pos] in ['O', '@']:
        work_list.append(grid[rel_pos])
        rel_pos = move(rel_pos, dir)
    if rel_pos in grid and grid[rel_pos] == '.':
        work_list = ['.'] + work_list
        rel_pos = pos
        for value in work_list:
            grid[rel_pos] = value
            rel_pos = move(rel_pos, dir)
        pos = move(pos, dir)
gps_total = 0
for pos, value in grid.items():
    top, left = pos
    if value == 'O':
        gps_total += 100 * top + left
print(gps_total)
    
```

```python
grid_master2, grid = [], {}
for line in grid_master:
    new_line = ''
    for elem in line:
        match elem:
            case '#':
                new_line += '##'
            case 'O':
                new_line += '[]'
            case '.':
                new_line += '..'
            case '@':
                new_line += '@.'
    grid_master2.append(new_line)
for row, line in enumerate(grid_master2):
    for col, value in enumerate(line):
        if value == '@':
            start_pos = (row, col)
        grid[(row, col)] = value
max_rows, max_cols = len(grid_master2), len(grid_master2[0])
```

```python
pos = start_pos
for dir in movements:
    row, col = pos
    if dir in ['<', '>']:
        rel_pos, work_list = pos, []
        while rel_pos in grid and grid[rel_pos] in ['[', ']', '@']:
            work_list.append(grid[rel_pos])
            rel_pos = move(rel_pos, dir)
        if rel_pos in grid and grid[rel_pos] == '.':
            work_list = ['.'] + work_list
            rel_pos = pos
            for value in work_list:
                grid[rel_pos] = value
                rel_pos = move(rel_pos, dir)
            pos = move(pos, dir)
    else:
        move_list = get_part2_coords(grid, dir)
        if {grid[move(x, dir)] for x in move_list}.issubset({'.', '[', ']'}) or not move_list:
            for elem in move_list:
                grid[elem] = '.'
            while move_list:
                grid[move(move_list.pop(0), dir)] = '['
                grid[move(move_list.pop(0), dir)] = ']'
            if grid[move(pos, dir)] != '#':
                grid[pos] = '.'
                grid[move(pos, dir)] = '@'
                pos = move(pos, dir)
gps_list = []
for loc in grid:
    if grid[loc] == '[':
        top, col = loc
        gps_list.append(100 * top + col)   
print(sum(gps_list))     
```
