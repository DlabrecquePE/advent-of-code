```python
import os
```

```python
def safe(rep):
    sorted_rep = sorted(rep)
    if sorted_rep == rep or sorted_rep == rep[::-1]:
        rep_diff = {rep[i + 1] - rep[i] for i in range(len(rep) - 1)}
        return True if rep_diff.issubset({1, 2, 3, -1, -2, -3}) else False
```

```python
file = 'data/day02.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
reports = []
for line in lines:
    reports.append([int(x) for x in line.split(' ')])
```

```python
# part one
safe_total = 0
for report in reports:
    if safe(report):
        safe_total += 1
print(safe_total)
```

```python
# part two
safe_total = 0
for report in reports:
    for i in range(len(report)):
        rep_copy = report.copy()
        del rep_copy[i]
        if safe(rep_copy):
            safe_total += 1
            break
print(safe_total)
```
