```python
import os
from collections import Counter
```

```python
file = 'data/test.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip() for x in fp.readlines()]
```

```python
def get_adjacent(grid, pos):
    x, y = pos
    adjacent = [(x - 1, y), (x + 1, y), (x, y - 1), (x, y + 1)]
    return {loc for loc in adjacent if loc in grid and grid[loc] != '#'}
```

```python
def get_shortest_path(grid, start, end) -> int:
    process, score, visited = {start}, 1, set()
    while process:
        new_process = set()
        for pos in process:
            for loc in get_adjacent(grid, pos).difference(visited):
                if loc == end:
                    return score
                else:
                    new_process.add(loc)
            visited.add(pos)
        score += 1
        process = new_process
    return 0
```

```python
grid = {}
for row, line in enumerate(lines):
    for col, elem in enumerate(line):
        if elem == 'S':
            start = (row, col)
        if elem == 'E':
            end = (row, col)
        grid[(row, col)] = elem
max_rows, max_cols = len(lines), len(lines[0])
```

```python
cheat_dict = Counter()
for row in range(max_rows):
    for col in range(max_cols):
        start = (row, col)
        if grid[start] != '.': 
            continue
        end = (row + 2, col)
        if (a := get_shortest_path(grid, start, end)):
            cheat_dict[a] += 1
        end = (row, col + 2)
        if (a := get_shortest_path(grid, start, end)):
            cheat_dict[a] += 1
print(cheat_dict)
```
