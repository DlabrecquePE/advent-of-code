```python
def guard_tick(guard_map, pos_heading, force_collision=False):
    row_old, col_old, heading = pos_heading
    max_rows, max_cols = len(guard_map), len(guard_map[0])
    check_dict = {0: (row_old - 1, col_old), 1: (row_old, col_old + 1), 
                  2: (row_old + 1, col_old), 3: (row_old, col_old - 1)}   
    row, col = check_dict[heading]
    if 0 <= row < max_rows and 0 <= col < max_cols:
        if guard_map[(row, col)] == '#' or force_collision:
            row, col, heading = row_old, col_old, (heading + 1) % 4
    return row, col, heading
```

```python
grid = []
for line in lines:
    grid.append(list(line))
max_rows, max_cols = len(grid), len(grid[0])
grid_iter = product(range(max_rows), range(max_cols))
grid_indices = set(grid_iter)
grid = np.array(grid)
```

```python
start_row, start_col = np.where(grid == '^')
row, col, heading, visited = *start_row, *start_col, 0, set()
while (row, col) in grid_indices:
    visited.add((row, col))
    row, col, heading = guard_tick(grid, (row, col, heading))
print(len(visited))
```

```python
obstructions = set()
for i, pos in enumerate(visited):
    if grid[pos] != '#' and pos != (start_row, start_col):
        new_grid = grid.copy()
        new_grid[pos], row, col, heading = '#', *start_row, *start_col, 0
        states = set()
        while (row, col) in grid_indices:
            if (row, col, heading) in states:
                obstructions.add(pos)
                break
            states.add((row, col, heading))
            row, col, heading = guard_tick(new_grid, (row, col, heading))
print(len(obstructions))
```
