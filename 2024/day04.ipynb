```python
import os
import numpy as np
```

```python
file = 'data/day04.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [list(x.rstrip()) for x in fp.readlines()]
```

```python
# part one
grid, iter, xmas_count = np.array(lines), len(lines), 0
for x in range(iter):
    d1, d2 = grid.diagonal(x - iter), grid.diagonal(x)
    d3, d4 = np.fliplr(grid).diagonal(x - iter), np.fliplr(grid).diagonal(x)
    test_lists = [grid[x, :], grid[:, x], d1, d2, d3, d4]
    test_lists.extend([np.flip(l) for l in test_lists])
    for test in test_lists:
        while len(test) > 0:
            if list(test[:4]) == ['X', 'M', 'A', 'S']:
                xmas_count += 1
            test = test[1:]
print(xmas_count)
```

```python
# part two
xmas_count = 0
for r in range(iter - 2):
    for c in range(iter - 2):
        if grid[r + 1, c + 1] == 'A':
            a = {grid[r, c], grid[r+2, c+2]} 
            b = {grid[r+2, c], grid[r, c+2]}
            if a == b == {'M', 'S'}:
                xmas_count += 1
print(xmas_count)
```
