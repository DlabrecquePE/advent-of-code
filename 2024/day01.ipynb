```python
import os
```

```python
file = 'data/day01.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
parse = [line.split('   ') for line in lines]
left = sorted([int(line[0]) for line in parse])
right = sorted([int(line[1]) for line in parse])
print(sum([abs(left[i] - right[i]) for i in range(len(left))]))
```

```python
# part two
print(sum([number * right.count(number) for number in left]))
```
