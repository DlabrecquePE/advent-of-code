```python
def trailheads(grid_dict: dict) -> list:
    elevation_target = 0
    paths = [[key] for key, value in grid.items() if value == 0]
    while True:
        new_paths = []
        elevation_target += 1
        for path in paths:
            x, y = path[-1]
            directions = [(x - 1, y), (x + 1, y), (x, y - 1), (x, y + 1)]
            for dir in directions:
                if dir in grid and grid[dir] == elevation_target:
                    new_paths.append(path + [dir])
        paths = new_paths
        if elevation_target == 9:
            return paths
```

```python
grid = {}
for row, line in enumerate(lines):
    for col, elevation in enumerate(line):
        grid[(row, col)] = int(elevation)
```
