```python
import os
```

```python
file = 'data/day07.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
test_dict = {}
for line in lines:
    result, parse = line.split(':')
    test_dict[int(result)] = [int(x) for x in parse.split()]
```

```python
# part one
total_calibration = 0
for test in test_dict:
    nums = test_dict[test]
    tree = {nums[0]}
    for num in nums[1:]:
        new = {i + num for i in tree if i + num <= test}
        new.update(j * num for j in tree if j * num <= test)
        tree = new
    if test in tree:
        total_calibration += test
print(total_calibration)
```

```python
# part two
total_calibration = 0
for test in test_dict:
    nums = test_dict[test]
    tree = {nums[0]}
    for num in nums[1:]:
        l = str(num)
        new = {i + num for i in tree if i + num <= test}
        new.update(j * num for j in tree if (j * num <= test))
        new.update(int(str(k) + l) for k in tree if int(str(k) + l) <= test)
        tree = new
    if test in tree:
        total_calibration += test
print(total_calibration)
```
