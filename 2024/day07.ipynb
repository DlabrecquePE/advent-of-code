```python
import os
```

```python
file = 'data/day07.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
test_dict = {}
for line in lines:
    result, parse = line.split(':')
    test_dict[int(result)] = [int(x) for x in parse.split()]
```

```python
# part one
calibration = 0
for test in test_dict:
    tree = {test_dict[test][0]}
    for elem in test_dict[test][1:]:
        tree = set([i + elem for i in tree] + [j * elem for j in tree])
    if test in tree:
        calibration += test
print(calibration)
```

```python
# part two
calibration = 0
for test in test_dict:
    tree = {test_dict[test][0]}
    for elem in test_dict[test][1:]:
        new_tree = set([i + elem for i in tree] + [j * elem for j in tree])
        new_tree.update(int(str(k) + str(elem)) for k in tree)
        tree = new_tree
    if test in tree:
        calibration += test
print(calibration)
```
