```python
import os
from itertools import product
from collections import Counter
```

```python
file = 'data/day14.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip() for x in fp.readlines()]
```

```python
bots, width, height, duration = [], 101, 103, 100
for line in lines:
    pos, vel = line.strip('p=').split(' v=')
    pos = tuple(map(int, pos.split(',')))
    vel = tuple(map(int, vel.split(',')))
    bots.append((pos, vel))
grid = {(x, y) for x, y in product(range(width), range(height))}
```

```python
def print_bots(bins):
    for num in bins:
        b_strs = [' ' if x == '0' else x for x in str(bin(num)[2:].zfill(width))]
        print(''.join(b_strs))
```

```python
def get_bins(bots):
    """ this was added as a misguided attempt to hash and store map
        states in an attempt to detect duplicates."""
    bins = []
    for y in range(height):
        b_str = ''.join(['1' if (x, y) in b_pos else '0' for x in range(width)])
        bins.append(int(b_str, 2))
    return tuple(bins)
```

```python
# part one
for i in range(duration):
    new_bots, b_pos = [], set()
    for bot in bots:
        pos, vel = bot
        p_x, p_y, v_x, v_y = *pos, *vel
        if (new_pos := (p_x + v_x, p_y + v_y)) not in grid:
            x, y = new_pos
            if x >= width or x < 0:
                x = width + x if x < 0 else x - width
            if y >= height or y < 0:
                y = height + y if y < 0 else y - height
            new_pos = (x, y)
        new_bots.append((new_pos, vel))
        b_pos.add(new_pos)
    bots = new_bots

quad_x, quad_y, cost_dict = width // 2, height // 2, Counter()
for bot in bots:
    x, y = bot[0]
    if x < quad_x and y < quad_y:
        cost_dict[0] += 1
    if x > quad_x and y < quad_y:
        cost_dict[1] += 1
    if x < quad_x and y > quad_y:
        cost_dict[2] += 1
    if x > quad_x and y > quad_y:
        cost_dict[3] += 1
total = 1
for cost in cost_dict.values():
    total *= cost
print(total)
```
