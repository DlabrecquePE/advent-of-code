```python
import os
```

```python
file = 'data/day13.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip() for x in fp.readlines()]
```

```python
def get_price(claw):
    a, b, c, d, e, f = claw
    price = 0
    if (x := (c*f - d*e) // (c*b - d*a)) == (c*f - d*e) / (c*b - d*a):
        if (y := (f - x*b) // d) == (f - x*b) / d:
            price += y + 3*x
    return price
```

```python
claws, i = [], 0
while lines:
    e, f = lines.pop().lstrip('Prize: X=').split(', Y=')
    c, d = lines.pop().lstrip('Button B: X').split(', Y')
    a, b = lines.pop().lstrip('Button A: X').split(', Y')
    if lines:
        lines.pop()
    claws.append([int(a), int(b), int(c), int(d), int(e), int(f)])
    i += 1
```

```python
tokens = 0
for claw in claws:
    tokens += get_price(claw)
print(tokens)     
```

```python
tokens = 0
for claw in claws:
    claw[-2] += 10000000000000
    claw[-1] += 10000000000000
    tokens += get_price(claw)
print(tokens) 
```
