```python
import os
from collections import Counter
```

```python
file = 'data/day11.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    line = [x.strip() for x in fp.readlines()]
```

```python
def stone_evolve(stones, blinks):
    for _ in range(blinks):
        new_stones = Counter()
        for stone, n in stones.items():
            split_string = str(stone)
            if not len(split_string) % 2:
                split_index = len(split_string) // 2
                right = (int(split_string[:split_index]))
                left = (int(split_string[split_index:]))
                new_stones[right] += n
                new_stones[left] += n
            else:
                new_stones[stone * 2024 if stone else 1] += n
        stones = new_stones
    return stones
```

```python
# part one
stones = Counter(int(x) for x in line[0].split())
print(sum(stone_evolve(stones, 25).values()))
```

```python
# part two
print(sum(stone_evolve(stones, 75).values()))
```
