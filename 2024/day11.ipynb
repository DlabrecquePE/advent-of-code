```python
import os
from collections import defaultdict
```

```python
file = 'data/day11.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    line = [x.strip() for x in fp.readlines()]
```

```python
def stone_evolve(stones):
    for _ in range(25):
        new_stones = defaultdict(int)
        for stone, n in stones.items():
            x, exp = stone
            split_string = str(x * 2024 ** exp)
            if not len(split_string) % 2:
                split_index = len(split_string) // 2
                right = (int(split_string[:split_index]), 0)
                left = (int(split_string[split_index:]), 0)
                new_stones[right] += n
                new_stones[left] += n
            else:
                new_stones[(x, exp + 1) if x else (1, 0)] += n
        stones = new_stones
    return stones
```

```python
# part one
stones = defaultdict()
stones.update({(int(x), 0): 1 for x in line[0].split()})
print(sum(stone_evolve(stones).values()))
```

```python
stones = defaultdict()
stones.update({(int(x), 0): 1 for x in line[0].split()})
stones = stone_evolve(stones)
stones = stone_evolve(stones)
print(sum(stone_evolve(stones).values()))
```
