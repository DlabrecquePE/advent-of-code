```python
import os
```

```python
file = 'data/day18.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip() for x in fp.readlines()]
```

```python
def get_grid(x_max, y_max, obstructions):
    grid = {}
    for x in range(x_max + 1):
        for y in range(y_max + 1):
            grid[(x, y)] = '#' if (x, y) in obstructions else '.'
    return grid
```

```python
def get_adjacent(grid, pos):
    x, y = pos
    adjacent = [(x - 1, y), (x + 1, y), (x, y - 1), (x, y + 1)]
    return {loc for loc in adjacent if loc in grid and grid[loc] != '#'}
```

```python
def print_grid(grid, x_max, y_max, highlight=[]):
    for x in range(x_max + 1):
        line = []
        for y in range(y_max + 1):
            line.append('O' if (x, y) in highlight else grid[(x, y)])
        print(''.join(line))
```

```python
def get_shortest_path(grid, start, end):
    process, score_dict, score = {start}, {}, 0
    while True:
        new_process = set()
        for pos in process:
            adjacent = get_adjacent(grid, pos)
            for loc in adjacent:
                if loc not in score_dict:
                    score_dict[loc] = score
                    new_process.add(loc)
        score += 1
        if new_process:
            process = new_process
        else:
            break
    if end not in score_dict:
        return 0
    path, pos = {end}, end
    while True:
        adjacent = list(get_adjacent(grid, pos))
        pos = min((score_dict[i], i) for i in adjacent)[1]
        if pos == start:
            return path
        path.add(pos)
```

```python
byte_list = []
for line in lines:
    y, x = line.split(',')
    byte_list.append((int(x), int(y)))
```

```python
# part one
elapsed, dim = 1024, 70
obstructions = byte_list[:elapsed]
grid = get_grid(dim, dim, obstructions)
start, end = (0, 0), (dim, dim)
shortest_path = get_shortest_path(grid, start, end)
print(len(shortest_path))
```

```python
path = shortest_path
for i, obstruction in enumerate(byte_list[elapsed:]):
    if obstruction not in path:
        continue
    grid = get_grid(dim, dim, byte_list[:elapsed + i + 1])
    if not (path := get_shortest_path(grid, start, end)):
        x, y = obstruction
        print(','.join([str(x), str(y)]))
        break
```

```python
index = offset = len(byte_list) // 2
while True:
    path = get_shortest_path(get_grid(dim, dim, byte_list[:index]), start, end)
    if not (offset := round(offset / 2)):
        break
    index += offset if path else -offset
print(','.join(map(str, byte_list[index])))
```
