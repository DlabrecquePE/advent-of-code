```python
import os
```

```python
file = 'data/day22.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip() for x in fp.readlines()]
```

```python
prune, total = 16777216, 0
for num in map(int, lines):
    for _ in range(2000):
        num = ((num << 6) ^ num) % prune
        num = ((num >> 5) ^ num) % prune
        num = ((num << 11) ^ num) % prune
    total += num
print(total)
```
