```python
import os
import re
```

```python
file = 'data/day03.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
pattern, line_master = re.compile('^mul\(\d{1,3},\d{1,3}\)'), ''.join(lines)
line, res = line_master, 0
while line:
    if pattern.match(line):
        x, y = line[4:line.index(')')].split(',')
        res += int(x) * int(y)
    line = line[1:]
print(res)
```

```python
# part two
line, res = line_master, 0
while line:
    if line.startswith("don't()"):
        while not line.startswith('do()'):
            line = line[1:]
    if pattern.match(line):
        x, y = line[4:line.index(')')].split(',')
        res += int(x) * int(y)
    line = line[1:]
print(res)
```
