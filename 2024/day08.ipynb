```python
import os
from collections import defaultdict
```

```python
file = 'data/day08.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
grid, antenna_dict = {}, defaultdict(list)
row_max, col_max = len(lines), len(lines[0])
for row, line in enumerate(lines):
    for col, loc in enumerate(line):
        grid[(row, col)] = loc
        if loc.isalnum():
            antenna_dict[loc].append((row, col))
```

```python
# part one
antinodes = set()
for antenna in antenna_dict:
    antenna_locs = set(antenna_dict[antenna])
    for loc in antenna_locs:
        loc_row, loc_col = loc
        for other_loc in antenna_locs.difference({loc}):
            if loc == other_loc:
                print(loc, other_loc)
                print(antenna_locs.difference(loc))
            other_row, other_col = other_loc
            antinode = (2 * loc_row - other_row, 2 * loc_col - other_col)
            if antinode in grid:
                antinodes.add(antinode)
print(len(antinodes))
```

```python
# part two
antinodes = set()
for antenna in antenna_dict:
    antenna_locs = set(antenna_dict[antenna])
    for loc in antenna_locs:
        loc_row, loc_col = loc
        for other_loc in antenna_locs.difference({loc}):
            other_row, other_col = other_loc
            row_diff, col_diff = loc_row - other_row, loc_col - other_col
            antinode = (loc_row - row_diff, loc_col - col_diff)
            while antinode in grid:
                antinodes.add(antinode)
                anti_row, anti_col = antinode
                antinode = (anti_row - row_diff, anti_col - col_diff)
print(len(antinodes))
```
