```python
import os
from collections import defaultdict
```

```python
file = 'data/day23.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip() for x in fp.readlines()]
```

```python
connections = defaultdict(list)
for line in lines:
    right, left = line.split('-')
    connections[right].append(left)
    connections[left].append(right)
```

```python
threeways, count = set(), 0
for elem1 in connections:
    for elem2 in connections[elem1]:
        for elem3 in connections[elem2]:
            if elem3 in connections[elem1]:
                threeways.add(frozenset({elem1, elem2, elem3}))
for threeway in threeways:
    if any(elem[0] == 't' for elem in threeway):
        count += 1
print(count)
```
