```python
import os
from collections import defaultdict
```

```python
file = 'data/test.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.strip() for x in fp.readlines()]
```

```python
towels = set()
while lines[0] != '':
    towels.update(lines.pop(0).split(', '))
patterns = [line for line in lines[1:]]
```

```python
def check_pattern(pattern, towel_list):
    remainders = [pattern]
    while remainders:
        remainder = remainders.pop()
        search = {remainder[:i + 1] for i in range(9)}.intersection(towel_list)
        for item in search:
            new_remainder = remainder[len(item):]
            if new_remainder:
                remainders.append(new_remainder)
            else:
                return True
    return False
```

```python
# part one
print(sum(check_pattern(pattern, towels) for pattern in patterns))
```

```python
def find_combos(pattern, towel_dict):
    check = {pattern[:i + 1] for i in range(9)}
    prefixes = check.intersection(towel_dict)
    res = []
    for prefix in prefixes:
        suffixes = find_combos(pattern[len(prefix):], towel_dict)
        print(prefix, suffixes)
        print('test', [prefix]+suffixes)
        res.extend([prefix] + suffix for suffix in suffixes)
        print(res)
    return res
```
