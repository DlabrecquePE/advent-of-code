```python
import os
from collections import Counter
```

```python
file = 'data/day02.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
valid_count = 0
for line in lines:
    ranges, char, password = line.split(' ')
    min_num, max_num = [int(x) for x in ranges.split('-')]
    char_count = Counter(password)
    if min_num <= char_count[char[0]] <= max_num:
        valid_count += 1
print(valid_count)
```

```python
# part two
valid_count = 0
for line in lines:
    ranges, char, password = line.split(' ')
    num1, num2 = [int(x) for x in ranges.split('-')]
    if (password[num1 - 1] == char[0]) ^ (password[num2 - 1] == char[0]):
        valid_count += 1
print(valid_count)
```
