```python
import os
from itertools import combinations
```

```python
file = 'data/day09.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
def find_invalid(sequence, preamble_length):
    for i, number in enumerate(sequence[preamble_length:]):
        preceding = set(sequence[i: i + preamble_length])
        sums = {i + j for i, j in combinations(preceding, 2)}
        if number not in sums:
            return number
```

```python
def find_weakness(sequence, invalid_number):
    width = 2
    while True:
        for i in range(len(sequence) - width):
            window = sequence[i: i + width]
            if sum(window) == invalid_number:
                return min(window) + max(window)
        width += 1
```

```python
# part one
sequence = [int(line) for line in lines]
invalid_number = find_invalid(sequence, 25)
print(invalid_number)
```

```python
# part two
print(find_weakness(sequence, invalid_number))
```
