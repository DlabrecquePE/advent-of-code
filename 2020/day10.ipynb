```python
import os
from collections import Counter
```

```python
file = 'data/test.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
adaps = sorted(int(line) for line in lines)
adaps = [0] + adaps + [max(adaps) + 3]
diffs = Counter(adaps[i + 1] - adaps[i] for i in range(len(adaps) - 1))
print(diffs[1] * diffs[3])
```

```python
links = {}
for adap in adaps:
    links[adap] = [x for x in adaps if x < adap <= x + 3]
link_tree = [[max(adaps)]]
print(links)
```
