```python
import os
```

```python
file = 'data/day06.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
count, data = 0, ''
for line in lines:   
    if line == '':
        count += len(set(data))
        data = ''
    else:
        data += line
print(count)
```

```python
# part two
count, data_list = 0, []
for line in lines:
    if line == '':
        count += len(set.intersection(*data_list))
        data_list = []
    else:
        data_list.append(set(line))
print(count)
```
