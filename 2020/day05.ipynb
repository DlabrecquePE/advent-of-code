```python
import os
```

```python
file = 'data/day05.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
class BoardingPass:
    def __init__(self, p_str: str): 
        self.row = int(p_str[:7].replace('F', '0').replace('B', '1'), 2)
        self.col = int(p_str[7:].replace('L', '0').replace('R', '1'), 2)
        self.seat_id = self.row*8 + self.col
```

```python
# part one
print(max(BoardingPass(line).seat_id for line in lines))
```

```python
# part two
id_list = [BoardingPass(line).seat_id for line in lines]
print(set(range(min(id_list), max(id_list))).difference(id_list).pop())
```
