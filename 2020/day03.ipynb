```python
import os
```

```python
file = 'data/day03.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
def tree_count(tree_map, slope):
    width, length, tree_count = len(lines[0]), len(lines), 0
    right, down, x, y = *slope, 0, 0
    while True:
        x += right
        y += down
        if y >= length:
            break
        if lines[y][x % width] == '#':
            tree_count += 1
    return tree_count   
```

```python
# part one
slope = (3, 1)
print(tree_count(lines, slope))  
```

```python
# part two
slope_list = [(1, 1), (3, 1), (5, 1), (7, 1), (1, 2)]
tree_mult = 1
for slope in slope_list:
    tree_mult *= tree_count(lines, slope)
print(tree_mult)
```
