```python
import os
from collections import defaultdict
```

```python
file = 'data/day07.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
bag_dict = {}
for line in lines:
    parse, bags = line.replace('bag,', 'bags,').replace('bag.', 'bags.'), []
    bag, parse = parse.removesuffix(' bags.').split(' bags contain ')
    if parse != 'no other':
        parse = parse.split(' bags, ')
        bags = [elem.split(' ', 1) for elem in parse]
        bags = {key: int(value) for value, key in bags}
    bag_dict[bag] = bags
```

```python
# part one
target_bag = 'shiny gold'
contain_count = 0
for bag in bag_dict:
    inner_bags = list(bag_dict[bag])
    while inner_bags:
        inner_bag = inner_bags.pop()
        if inner_bag == target_bag:
            contain_count += 1
            break
        inner_bags.extend(bag_dict[inner_bag])
print(contain_count)
```

```python
# part two
contents_dict = defaultdict(int)
while bag_dict:
    leaves = {bag for bag in bag_dict if not bag_dict[bag]}
    for leaf in leaves:
        contents_dict[leaf] += 1
        del bag_dict[leaf]
        for bag in bag_dict:
            if leaf in bag_dict[bag]:
                contents_dict[bag] += contents_dict[leaf] * bag_dict[bag][leaf]
                del bag_dict[bag][leaf]
print(contents_dict[target_bag] - 1)
```
