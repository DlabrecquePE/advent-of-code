```python
import os
```

```python
file = 'data/day04.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
passport_list, passport_string = [], ''
for line in lines:
    if line == '':
        passport_list.append(passport_string)
        passport_string = ''
    else:
        passport_string = f'{passport_string} {line}'.lstrip()
passport_dict = {}
for key, passport in enumerate(passport_list):
    data_list = list(passport.split(' '))
    data_dict = {i: j for i, j in [k.split(':') for k in data_list]}
    passport_dict[key] = data_dict
```

```python
field_complete = {'byr', 'iyr', 'eyr', 'hgt', 'hcl', 'ecl', 'pid', 'cid'}
field_exclude = {'cid'}
field_include = field_complete - field_exclude
```

```python
# part one
valid_count = 0
for passport in passport_dict.values():
    if field_include.issubset(passport):
        valid_count += 1
print(valid_count)
```

```python
# part two
dig = {'1', '2', '3', '4', '5', '6', '7', '8', '9', '0'}
h_dig = dig.union({'a','b','c','d','e','f'})
eye_colors = {'amb', 'blu', 'brn', 'gry', 'grn', 'hzl', 'oth'}
valid_count = 0
for passport in passport_dict.values():
    if field_include.issubset(passport):
        byr = passport['byr']
        checks = [dig.issuperset(byr) and 1920 <= int(byr) <= 2002]
        iyr = passport['iyr']
        checks.append((dig.issuperset(iyr) and 2010 <= int(iyr) <= 2020))
        eyr = passport['eyr']
        checks.append((dig.issuperset(eyr) and 2020 <= int(eyr) <= 2030))
        hgt = passport['hgt']
        hgt, hgt_units = hgt[:-2], hgt[-2:]
        match hgt_units:
            case 'cm':
                checks.append((dig.issuperset(hgt) and 150 <= int(hgt) <= 193))
            case 'in':
                checks.append((dig.issuperset(hgt) and 59 <= int(hgt) <= 76))
            case _:
                checks.append(False)
        hcl = passport['hcl']
        pre, hcl = hcl[0], hcl[1:]
        checks.append((pre == '#' and len(hcl) == 6 and h_dig.issuperset(hcl))) 
        ecl = passport['ecl']
        checks.append(ecl in eye_colors)
        pid = passport['pid']
        checks.append((dig.issuperset(pid) and len(pid) == 9))
        if all(checks):
            valid_count += 1
print(valid_count)
```
