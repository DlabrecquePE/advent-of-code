```python
import os
```

```python
file = 'data/day01.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
input, target = [int(line) for line in lines], 2020
while input:
    test = input.pop()
    for other in input:
        if test + other == target:
            print(test * other)
            break
        
```

```python
# part two
num_list, target = [], 2020
input = [int(line) for line in lines]
for i, a in enumerate(input[:-1]):
    for j, b in enumerate(input[i:]):
        for k, c in enumerate(input[j:]):
            if i != j and j != k and k != i and a + b + c == target:
                print(a * b * c)
                break
```
