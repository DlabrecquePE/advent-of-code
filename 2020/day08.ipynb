```python
import os
```

```python
file = 'data/day08.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
def loop_test(instructions: list) -> tuple[str, int]:
    loop_set, inst_pointer, acc = {len(instructions) - 1}, 0, 0
    while True:
        inst, arg = instructions[inst_pointer]
        if inst_pointer in loop_set:
            return (inst, acc)
        loop_set.add(inst_pointer)
        if inst == 'acc':
            acc += arg
        inst_pointer += arg if inst == 'jmp' else 1
```

```python
def gen_program(instructions):
    for i, inst in enumerate(instructions):
        if inst[0] == 'nop':
            yield instructions[:i] + [('jmp', inst[1])] + instructions[i + 1:]
        if inst[0] == 'jmp':
            yield instructions[:i] + [('nop', inst[1])] + instructions[i + 1:]
```

```python
# part one
inst_list = [(i, int(j)) for i, j in [k.split() for k in lines]]
print(loop_test(inst_list)[1])
```

```python
# part two
inst_list.append(('halt', 0))
progger = gen_program(inst_list)
for prog in progger:
    last_inst, acc = loop_test(prog)
    if last_inst == 'halt':
        break
print(acc)
```
