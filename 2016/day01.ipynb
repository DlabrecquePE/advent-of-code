```python
import os
```

```python
file = 'data/day01.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    line = fp.readlines()
instructions = [x.strip() for x in line[0].split(',')]
```

```python
instructions = ['R8', 'R4', 'R4', 'R8']
loc = [0, 0]
heading = 'N'
locs = []
for instruction in instructions:
    turn = instruction[0]
    blocks = int(instruction[1:])
    old_loc = loc.copy()
    match heading:
        case 'N':
            heading = 'W' if turn == 'L' else 'E'
            loc[1] += blocks if heading == 'E' else -blocks
        case 'S':
            heading = 'E' if turn == 'L' else 'W'
            loc[1] += blocks if heading == 'E' else -blocks
        case 'E':
            heading = 'N' if turn == 'L' else 'S'
            loc[0] += blocks if heading == 'N' else -blocks
        case 'W':
            heading = 'S' if turn == 'L' else 'N'
            loc[0] += blocks if heading == 'N' else -blocks
    start_x, start_y = old_loc
    end_x, end_y = loc
    step = -1 if start_x > end_x or start_y > end_y else 1
    loc_min = start_x if start_x != end_x else start_y
    loc_max = end_x if end_x != start_x else end_y
    print(start_x, start_y, end_x, end_y, loc_min, loc_max, step)
    for pos in range(loc_min, loc_max + 1, step):
        locs.append((loc[0], pos) if heading == 'W' or 'E' else (pos, loc[1]))
    print(loc)
print('Part One:', abs(loc[0]) + abs(loc[1]))
for i, loc in enumerate(locs):
    if loc in locs[:i]:
        print('Part Two:', abs(loc[0]) + abs(loc[1]))
        break
```
