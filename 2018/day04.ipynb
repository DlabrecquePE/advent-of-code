```python
import os
from collections import Counter
```

```python
file = 'data/day04.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
log_dict, guard_dict = {}, {}
for line in lines:
    timestamp, log = line.strip('[').split('] ')
    date, time = timestamp.split()
    year, month, day = map(int, date.split('-'))
    hour, minute = map(int, time.split(':'))
    log_dict[(year, month, day, hour, minute)] = log
for timestamp in sorted(log_dict.keys()):
    hour, minute, log = timestamp[3], timestamp[4], log_dict[timestamp]
    if log.startswith('Guard #'):
        guard = int(log.lstrip('Guard #').rstrip(' begins shift'))
        if guard not in guard_dict:
            guard_dict[guard] = Counter()
    elif log == 'falls asleep':
        start = minute
    else:
        guard_dict[guard].update(list(range(start, minute)))
```

```python
# part one
minutes_dict = {}
for guard in guard_dict:
    guard_counter = guard_dict[guard]
    if guard_counter:
        total_minutes = sum(guard_counter.values())
        freq_minute = max(guard_counter, key=guard_counter.get)
        freq_minute_count = guard_counter[freq_minute]
        minutes_dict[guard] = (total_minutes, freq_minute, freq_minute_count)
sleepiest_guard = max(minutes_dict, key=minutes_dict.get)
sleepiest_guard_max_minute = minutes_dict[sleepiest_guard][1]
print(sleepiest_guard * sleepiest_guard_max_minute)
```

```python
# part two
minutes_dict = {k: (v[2], v[1]) for k, v in minutes_dict.items()}
sleepiest_minute = max(minutes_dict.values())[1]
sleepiest_minute_guard = max(minutes_dict, key=minutes_dict.get)
print(sleepiest_minute * sleepiest_minute_guard)
```
