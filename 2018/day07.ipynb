```python
import os
from collections import defaultdict
```

```python
file = 'data/day07.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
step_dict_master, all_steps = defaultdict(set), set()
for line in lines:
    parse = line.removeprefix('Step ').removesuffix(' can begin.')
    pre_step, step = parse.split(' must be finished before step ')    
    step_dict_master[step].add(pre_step)
    all_steps.update([step, pre_step])
```

```python
# part one
step_dict, step_order = dict(step_dict_master), []
obstructed, unfinished = sorted(list(step_dict.keys())), all_steps.copy()
while unfinished:
    unobstructed = unfinished.difference(obstructed)
    step = sorted(list(unobstructed))[0]
    unfinished.remove(step)
    step_order.append(step)
    step_dict = {k: {i for i in step_dict[k] if i != step} for k in step_dict}
    step_dict = {k: v for k, v in step_dict.items() if v}
    obstructed = sorted(list(step_dict.keys()))
print(''.join(step_order))
```

```python
# part two
workers, work_dict, time, time_offset = 5, {}, 0, 60
step_dict, step_order = dict(step_dict_master), []
unfinished = all_steps.copy()
while unfinished or work_dict:
    obstructed, in_progress = step_dict.keys(), work_dict.keys()
    unobstructed = unfinished.difference(obstructed).difference(in_progress)
    while len(work_dict) < workers and unobstructed:
        step = sorted(list(unobstructed))[0]
        work_dict[step] = ord(step) - 64 + time_offset
        unobstructed.remove(step)
    step_time = min(work_dict.values())
    time += step_time
    work_dict = {k: v - step_time for k, v in work_dict.items()}
    for elem in work_dict.copy():
        if work_dict[elem] == 0:
            step_order += elem
            del work_dict[elem]
            temp = {k: v.difference(elem) for k, v in step_dict.items()}
            step_dict = {k: v for k, v in temp.items() if v}
            unfinished.remove(elem)
print(time)
```
