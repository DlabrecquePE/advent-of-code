```python
import os
```

```python
file = 'data/day05.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    line = fp.readline()
```

```python
def react_polymer(poly):
    new_poly = [poly[0]]
    for unit in poly[1:]:
        if new_poly and abs(ord(unit) - ord(new_poly[-1])) == 32:
            new_poly.pop()
        else:
            new_poly.append(unit)
    return ''.join(new_poly[:-1])
```

```python
# part one
print(len(react_polymer(line)))
```

```python
# part two
poly_lens = []
for x in range(65, 91):
    test_poly = line.replace(chr(x), '').replace(chr(x + 32), '')
    poly_lens.append(len(react_polymer(test_poly)))
print(min(poly_lens))
```
