```python
import os
from itertools import cycle
```

```python
file = 'data/day01.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
print(sum([int(elem) for elem in lines]))
```

```python
# part two
freq = 0
freqs = {freq}
for line in cycle(lines):
    freq += int(line)
    if freq not in freqs:
        freqs.add(freq)
    else:
        break
print(freq)
    
```
