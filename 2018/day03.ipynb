```python
import os
from itertools import product
```

```python
file = 'data/day03.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
claim_dict = {}
for line in lines:
    claim_id, line = line.lstrip('#').split(' @ ')
    left , line = line.split(',')
    top, line = line.split(': ')
    width, height = line.split('x')
    claim_dict[int(claim_id)] = [int(left), int(top), int(width), int(height)]
```

```python
# part one
claimed_coords, overlap_coords = set(), set()
for claim in claim_dict:
    l, t, w, h = claim_dict[claim]
    l_range, t_range = list(range(l, l + w)), list(range(t, t + h))
    for coord in product(l_range, t_range):
        if coord in claimed_coords:
            overlap_coords.add(coord)
        else:
            claimed_coords.add(coord)
print(len(overlap_coords))
```

```python
# part two
for claim in claim_dict:
    l, t, w, h = claim_dict[claim]
    l_range, t_range = list(range(l, l + w)), list(range(t, t + h))
    for coord in product(l_range, t_range):
        if coord in overlap_coords:
            break
    else:
        break
print(claim)
```
