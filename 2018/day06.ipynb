```python
import os
from itertools import product
from collections import Counter
```

```python
file = 'data/day06.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
    lines = [line.split(',') for line in lines]
    lines = [(int(x), int(y)) for x, y in lines]
```

```python
x_max, y_max = [max(elem) for elem in list(zip(*lines))]
x_range, y_range = range(x_max + 1), range(y_max + 1)
```

```python
# part one
grid, grid_iter = {}, product(x_range, y_range)
for coord in grid_iter:
    x, y = coord
    dist_dict = {i: abs(x - v[0]) + abs(y - v[1]) for i, v in enumerate(lines)}
    distances = dist_dict.values()
    if Counter(distances)[min(distances)] == 1:
        grid[coord] = min(dist_dict, key=dist_dict.get)

border_coords = [(0, y) for y in y_range] + [(x, 0) for x in x_range]
border_coords += [(x_max, y) for y in y_range] + [(x, y_max) for x in x_range]
init_elim = set(grid[coord] for coord in border_coords if coord in grid)
areas = Counter([grid[coord] for coord in grid if grid[coord] not in init_elim])
print(max(areas.values()))
```

```python
# part two
max_dist, grid_iter, area = 10000, product(x_range, y_range), 0
for coord in grid_iter:
    x, y = coord
    total_distance = sum([abs(x - v[0]) + abs(y - v[1]) for v in lines])
    if total_distance < max_dist:
        area += 1
print(area)
```
