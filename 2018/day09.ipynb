```python
import os
from collections import defaultdict, deque
```

```python
file = 'data/day09.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    line = fp.readline()
line = line.strip(' points\n').split(' players; last marble is worth ')
```

```python
def marble_game(players, marbles):
    score_dict, circle = defaultdict(int), deque([0])
    for marble in range(1, marbles + 1):
        if not marble % 23:
            circle.rotate(7)
            score_dict[marble % players] += marble + circle.pop()
            circle.rotate(-1)
        else:
            circle.rotate(-1)
            circle.append(marble)
    return max(score_dict.values())
```

```python
# part one
p, m = [int(string) for string in line]
print(marble_game(p, m))
```

```python
# part two
p, m = [int(string) for string in line]
print(marble_game(p, m * 100))
```
