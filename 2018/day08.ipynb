```python
import os
```

```python
file = 'data/day08.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    line = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
tree, meta_total = [int(x) for x in line[0].split(' ')], 0
while tree:
    begin = tree.index(0)
    meta_count = tree[begin + 1]
    end = begin + meta_count + 2
    meta_slice = tree[begin + 2: end]
    meta_total += sum(meta_slice)
    tree[begin - 2] -= 1
    tree = tree[:begin] + tree[end:]
print(meta_total)
```
