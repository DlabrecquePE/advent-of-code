```python
import os
from collections import Counter
from itertools import combinations
```

```python
file = 'data/day02.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
pairs, triples = 0, 0
for line in lines:
    char_count = Counter(line).values()
    if 2 in char_count:
        pairs += 1
    if 3 in char_count:
        triples += 1
print(pairs * triples)
```

```python
target_len = len(lines[0]) - 1
for a, b in combinations(lines, 2):
    common_letters = [x for x, y in zip(a, b) if x == y]
    if len(common_letters) == target_len:
        break
print(''.join(common_letters))
```
