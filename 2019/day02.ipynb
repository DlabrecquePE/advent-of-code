```python
import os
```

```python
def intcode_interpreter(program):
    current_pos = 0
    while True:
        opcode = program[current_pos]
        if opcode == 99:
            return program[0]
        noun_pos, verb_pos, out_pos = program[current_pos + 1: current_pos + 4]
        noun, verb = program[noun_pos], program[verb_pos]
        match opcode:
            case 1:
                program[out_pos] = noun + verb
                current_pos += 4
            case 2:
                program[out_pos] = noun * verb
                current_pos += 4
```

```python
file = 'data/day02.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    line = fp.readline()
```

```python
# part one
program = [int(x) for x in line.split(',')]
program[1], program[2] = 12, 2 # edit this line out for test data
print(intcode_interpreter(program))
```

```python
# part two
output_target = 19690720
for noun in range(100):
    for verb in range(100):
        program = [int(x) for x in line.split(',')]
        program[1], program[2] = noun, verb
        if intcode_interpreter(program) == output_target:
            print(100 * noun + verb)
```
