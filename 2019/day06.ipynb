```python
import os
from collections import defaultdict
```

```python
file = 'data/day06.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
orbit_tree, edge_dict = defaultdict(set), defaultdict(set)
for line in lines:
    parent, child = line.split(')')
    orbit_tree[parent].add(child)
    edge_dict[parent].add(child)
    edge_dict[child].add(parent)
```

```python
# part one
orbit_dict = {key: [0, 0] for key in edge_dict}
while orbit_tree:
    parents = set(orbit_tree.keys())
    children = set(flat for lists in orbit_tree.values() for flat in lists)
    leaves = children.difference(parents)
    for node in parents.union(children).difference(leaves):
        for leaf in leaves.intersection(orbit_tree[node]):
            orbit_dict[node][0] += 1
            orbit_dict[node][1] += sum(orbit_dict[leaf])
    orbit_tree = {k: v.difference(leaves) for k, v in orbit_tree.items()}
    orbit_tree = {k: v for k, v in orbit_tree.items() if v}
a, b = zip(*orbit_dict.values())
print(sum(a + b))
```

```python
# part two
current_node, target_node, step_count = 'YOU', 'SAN', 0
step_dict = {step_count: edge_dict[current_node]}
while True:
    next_steps = {step for step in step_dict[step_count]}
    if target_node in next_steps:
        break
    step_count += 1
    step_dict[step_count] = []
    for step in next_steps:
        step_dict[step_count].extend(edge_dict[step])
print(step_count - 1)
```
