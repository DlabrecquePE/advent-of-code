```python
import os
```

```python
file = 'data/day05.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    line = [x.rstrip() for x in fp.readlines()][0]
```

```python
def intcode_interpreter(prog, prog_input):
    inst_pointer = 0
    while True:
        inst = str(prog[inst_pointer])
        opcode, params = int(inst[-2:]), list(str(inst[-3::-1]))
        if opcode == 99:
            return prog
        match opcode:
            case 1: # add
                a_pos, b_pos = prog[inst_pointer+1], prog[inst_pointer+2]
                a_param = int(params.pop(0)) if params else 0
                b_param = int(params.pop(0)) if params else 0
                a = a_pos if a_param else prog[a_pos]
                b = b_pos if b_param else prog[b_pos]
                out = prog[inst_pointer+3]
                prog[out] = a + b
                inst_pointer += 4
            case 2: # mult
                a_pos, b_pos = prog[inst_pointer+1], prog[inst_pointer+2]
                a_param = int(params.pop(0)) if params else 0
                b_param = int(params.pop(0)) if params else 0
                a = a_pos if a_param else prog[a_pos]
                b = b_pos if b_param else prog[b_pos]
                out = prog[inst_pointer+3]
                prog[out] = a * b
                inst_pointer += 4
            case 3: # set addr to prog_input
                out = prog[inst_pointer+1]
                prog[out] = prog_input
                inst_pointer += 2
            case 4: # print addr as output
                a_pos = prog[inst_pointer+1]
                a_param = int(params.pop(0)) if params else 0
                a = a_pos if a_param else prog[a_pos]
                print('output:', a)
                inst_pointer += 2
            case 5: # jump-if-true
                a_pos, b_pos = prog[inst_pointer+1], prog[inst_pointer+2]
                a_param = int(params.pop(0)) if params else 0
                b_param = int(params.pop(0)) if params else 0
                a = a_pos if a_param else prog[a_pos]
                b = b_pos if b_param else prog[b_pos]
                if a != 0:
                    inst_pointer = b
                else:
                    inst_pointer += 3
            case 6: # jump-if-false
                a_pos, b_pos = prog[inst_pointer+1], prog[inst_pointer+2]
                a_param = int(params.pop(0)) if params else 0
                b_param = int(params.pop(0)) if params else 0
                a = a_pos if a_param else prog[a_pos]
                b = b_pos if b_param else prog[b_pos]
                if a == 0:
                    inst_pointer = b
                else:
                    inst_pointer += 3
            case 7: # less than
                a_pos, b_pos = prog[inst_pointer+1], prog[inst_pointer+2]
                a_param = int(params.pop(0)) if params else 0
                b_param = int(params.pop(0)) if params else 0
                a = a_pos if a_param else prog[a_pos]
                b = b_pos if b_param else prog[b_pos]
                out = prog[inst_pointer+3]
                prog[out] = 1 if a < b else 0
                inst_pointer += 4
            case 8: # equals
                a_pos, b_pos = prog[inst_pointer+1], prog[inst_pointer+2]
                a_param = int(params.pop(0)) if params else 0
                b_param = int(params.pop(0)) if params else 0
                a = a_pos if a_param else prog[a_pos]
                b = b_pos if b_param else prog[b_pos]
                out = prog[inst_pointer+3]
                prog[out] = 1 if a == b else 0
                inst_pointer += 4
```

```python
# part one
program = [int(x) for x in line.split(',')]
_ = intcode_interpreter(program, 1)
```

```python
# part two
program = [int(x) for x in line.split(',')]
_ = intcode_interpreter(program, 5)
```
