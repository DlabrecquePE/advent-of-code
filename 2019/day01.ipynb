```python
import os
```

```python
def fuel_req(num):
    return num // 3 - 2
```

```python
file = 'data/day01.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
total = 0
for line in lines:
    total += fuel_req(int(line))
print(total)
```

```python
# part two
total = 0
for line in lines:
    mass = int(line)
    fuel_add = fuel_req(mass)
    while fuel_add > 0:
        total += fuel_add
        fuel_add = fuel_req(fuel_add)
print(total)
```
