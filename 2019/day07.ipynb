```python
import os
from itertools import permutations
```

```python
file = 'data/day07.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
def intcode_interpreter(prog, *args):
    inst_pointer = 0
    while True:
        inst = str(prog[inst_pointer])
        opcode = int(inst[-2:])
        if opcode == 99:
            return output
        params = [int(x) for x in inst[:-2].rjust(3, '0')]
        next_inst = prog[inst_pointer + 1: inst_pointer + 4]
        a = next_inst[0] if params.pop() else prog[next_inst[0]]
        if opcode not in [3, 4]:
            b = next_inst[1] if params.pop() else prog[next_inst[1]]
        else:
            inst_pointer += 2
        if opcode in [1, 2, 7, 8]:
            c = next_inst[2]
            inst_pointer += 4
        match opcode:
            case 1: # add
                prog[c] = a + b
            case 2: # mult
                prog[c] = a * b
            case 3: # set addr to prog_input
                prog[next_inst[0]] = args[0]
                if len(args) > 1:
                    args = args[1:]
            case 4: # print addr as output
                output = prog[next_inst[0]]
            case 5: # jump-if-true
                inst_pointer = b if a != 0 else inst_pointer + 3
            case 6: # jump-if-false
                inst_pointer = b if a == 0 else inst_pointer + 3
            case 7: # less than
                prog[c] = 1 if a < b else 0
            case 8: # equals
                prog[c] = 1 if a == b else 0
```

```python
amp_phases = set()
for tup in permutations('01234', 5):
    amp_phases.add(tuple(int(x) for x in tup))
    # amp_phases.add(tuple(int(x) for x in tup)[::-1])
prog = [int(x) for x in lines[0].split(',')]
```

```python
# part one
signals = []
input = 0
for phases in amp_phases:
    output = intcode_interpreter(prog, phases[0], input)
    output = intcode_interpreter(prog, phases[1], output)
    output = intcode_interpreter(prog, phases[2], output)
    output = intcode_interpreter(prog, phases[3], output)
    output = intcode_interpreter(prog, phases[4], output)
    signals.append(output)
print(max(signals))
```
