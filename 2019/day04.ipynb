```python
import os
from collections import Counter
```

```python
file = 'data/day04.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    password_input = fp.readlines()
```

```python
# part one
begin, end = (int(x) for x in password_input[0].split('-'))
password_count = 0
for x in range(begin, end + 1):
    digits = str(x)
    doubles = any(digits[i] == digits[i+1] for i in range(5))
    ascending = all(digits[i] <= digits[i+1] for i in range(5))
    if doubles and ascending:
        password_count += 1
print(password_count)    
```

```python
# part two
password_count = 0
for x in range(begin, end + 1):
    digits = str(x)
    d_count = Counter(digits)
    doubles = [int(d) for d in d_count.keys() if d_count[d] == 2]
    ascending = all(digits[i] <= digits[i+1] for i in range(5))
    if doubles and ascending:        
        large_groups = [int(d) for d in d_count.keys() if d_count[d] > 2]
        if not large_groups or any(x for x in doubles if x not in large_groups):
            password_count += 1
print(password_count)
```
