```python
import os
```

```python
file = 'data/test.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
# part one
wire_dict = {}
for i, line in enumerate(lines):
    wire_dict[i] = line.split(',')
coord_dict = {0: [], 1: []}
for wire in wire_dict:
    current_x, current_y = 0, 0
    for path in wire_dict[wire]:
        direction = path[0]
        steps = int(path[1:])
        match direction:
            case 'R':
                for offset in range(1, steps + 1):
                    coord_dict[wire].append((current_x + offset, current_y))
                current_x += steps
            case 'L':
                for offset in range(1, steps + 1):
                    coord_dict[wire].append((current_x - offset, current_y))
                current_x -= steps
            case 'U':
                for offset in range(1, steps + 1):
                    coord_dict[wire].append((current_x, current_y + offset))
                current_y += steps
            case 'D':
                for offset in range(1, steps + 1):
                    coord_dict[wire].append((current_x, current_y - offset))
                current_y -= steps
intersections = set(coord_dict[0]).intersection(coord_dict[1])
distance_list = []
for coord in intersections:
    distance_list.append(abs(coord[0]) + abs(coord[1]))
print(min(distance_list))
```

```python
# part two
coord_dict = {0: [], 1: []}
for wire in wire_dict:
    current_x, current_y = 0, 0
    for path in wire_dict[wire]:
        direction = path[0]
        steps = int(path[1:])
        total_steps = 0
        coord_data = {}
        match direction:
            case 'R':
                for offset in range(1, steps + 1):
                    total_steps += 1
                    coord = (current_x + offset, current_y)
                    coord_data[coord] = total_steps
                if not coord_dict[wire][coord]:
                    coord_dict[wire] = coord_data
                current_x += steps
            case 'L':
                for offset in range(1, steps + 1):
                    total_steps += 1
                    coord = (current_x - offset, current_y)
                    coord_data[coord] = total_steps
                    if not coord_dict[wire][coord]:
                        coord_dict[wire].append(coord_data)
                current_x -= steps
            case 'U':
                for offset in range(1, steps + 1):
                    total_steps += 1
                    coord = (current_x, current_y + offset)
                    coord_data[coord] = total_steps
                    if not coord_dict[wire][coord]:
                        coord_dict[wire].append(coord_data)
                current_y += steps
            case 'D':
                for offset in range(1, steps + 1):
                    total_steps += 1
                    coord = (current_x, current_y - offset)
                    coord_data[coord] = total_steps
                    if not coord_dict[wire][coord]:
                        coord_dict[wire].append(coord_data)
                current_y -= steps
intersections = set(coord_dict[0].keys()).intersection(coord_dict[1].keys())
print(intersections)

```
