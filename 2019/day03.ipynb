```python
import os
```

```python
file = 'data/day03.txt'
path = os.path.join(os.getcwd(), file)
with open(path, 'r') as fp:
    lines = [x.rstrip() for x in fp.readlines()]
```

```python
wire_dict = {0: {}, 1: {}}
d_map = {'U': (0, 1), 'D': (0, -1), 'L': (-1, 0), 'R': (1, 0)}
for wire_number, line in enumerate(lines):
    instructions, delay = line.split(','), 0
    coord = (0, 0)
    grid_positions = {}
    for elem in instructions:
        direction, steps = elem[0], int(elem[1:])
        x, y = coord
        for i in range(steps):
            coord = tuple(x + y for x,y in zip(coord, d_map[direction]))
            delay += 1
            wire_dict[wire_number][coord] = delay
```

```python
# part one
intersects = set(wire_dict[0].keys()).intersection(wire_dict[1].keys())
print(min([(abs(x[0]) + abs(x[1])) for x in intersects]))
```

```python
# part two
print(min([(wire_dict[0][x] + wire_dict[1][x]) for x in intersects]))
```
